# III. Fundamento teórico
Un A/B test compara un control A con una variante B mediante asignación aleatoria. La unidad experimental de este caso es el usuario, que conserva su variante durante toda la observación. La aleatorización favorece la comparabilidad en expectativa; no garantiza grupos idénticos.

La respuesta es binaria: compra dentro de los siete días posteriores a la asignación. Todos los usuarios completan esa ventana antes del análisis. La métrica primaria es la proporción de usuarios que compran. Se estudia la diferencia de dos proporciones, no ANOVA.

El MDE es el efecto utilizado para planificar la potencia. El umbral empresarial es la mejora que justificaría interés práctico. Aquí son distintos: MDE de 1,5 pp y umbral empresarial de 1,0 pp. La muestra se dimensiona para detectar el MDE frente a cero, no para garantizar evidencia de superar el umbral empresarial.

Se usa una aproximación normal de Wald para la diferencia. El contraste unilateral al 5 % y el límite inferior unilateral del 95 % utilizan el mismo error estándar. También se muestra un IC bilateral del 95 % como descripción de incertidumbre; no se exige que este último excluya cero para reproducir el contraste unilateral.

La recomendación académica requiere evidencia de superar el umbral empresarial y de mantener el guardrail dentro del margen predefinido. Son criterios más exigentes que obtener p < 0,05 frente a cero. Un resultado no concluyente no demuestra igualdad.

### 2.3 Hipótesis y reglas previas
Conversión: H0: pB−pA ≤ 0; H1: pB−pA > 0. Alfa = 0,05 unilateral.

Relevancia empresarial: exigir límite inferior unilateral del 95 % mayor que 0,010. Esto evalúa un umbral distinto de cero y no se deduce solo de la estimación puntual.

Guardrail: H0: qB−qA ≥ 0,005; H1: qB−qA < 0,005. Exigir límite superior unilateral del 95 % menor que 0,005. No encontrar una diferencia significativa no demuestra que el guardrail esté protegido.

La recomendación combina ambos requisitos; se trata de una regla conjunta, no de escoger la prueba favorable. Los intervalos individuales no deben describirse como una región de confianza simultánea.

**Comprobación:** ¿por qué se comparan proporciones? ¿Cómo cambiarían las hipótesis si interesara cualquier diferencia, positiva o negativa?
**Respuesta:** Se comparan proporciones porque la variable respuesta es binaria y la métrica primaria representa la proporción de usuarios que compran. Si interesara cualquier diferencia, positiva o negativa, la hipótesis alternativa sería bilateral: H1: pB−pA ≠ 0, en lugar de H1: pB−pA > 0.

### Paso 2. Fijar los parámetros antes de observar
Los baselines son supuestos históricos del caso ficticio, no estimaciones obtenidas de los grupos que analizaremos. El margen del guardrail se fija por criterio de negocio. Se calcula muestra para conversión y, por aproximación normal, para no inferioridad del guardrail suponiendo diferencia verdadera cero. Se toma el mayor tamaño. Las potencias se refieren a cada objetivo de diseño; no garantizan 80 % para la regla conjunta ni para superar el umbral empresarial.


```python
# ============================================================
# CONFIGURACIÓN DE PARÁMETROS DEL EXPERIMENTO
# ============================================================

# Definir el nivel de significancia y la potencia estadística
ALPHA = 0.05  ### Nivel de significancia del 5 %
POTENCIA = 0.80  ### Potencia estadística del 80 %

# Definir el baseline de conversión y el efecto mínimo detectable
BASELINE = 0.118  ### Conversión histórica del caso: 11.8 %
MDE = 0.015  ### Efecto mínimo detectable: 1.5 puntos porcentuales

# Definir el umbral de negocio
UMBRAL_NEGOCIO = 0.010  ### Aumento mínimo considerado relevante para el negocio: 1.0 punto porcentual

# Definir el baseline y el margen permitido para el guardrail
BASE_GUARDRAIL = 0.018  ### Incidencia histórica del guardrail: 1.8 %
MARGEN_GUARDRAIL = 0.005  ### Deterioro máximo permitido: 0.5 puntos porcentuales

# Fijar la semilla para garantizar la reproducibilidad
SEMILLA = 20260921  ### Semilla utilizada para obtener los mismos resultados en la simulación

# Definir el periodo de reclutamiento y la ventana de seguimiento
DIAS_RECLUTAMIENTO = 14  ### Días destinados al reclutamiento de usuarios
VENTANA_DIAS = 7  ### Días de seguimiento para medir la compra

# ============================================================
# VALIDACIÓN DE LOS PARÁMETROS
# ============================================================

# Verificar que el baseline y el MDE produzcan una proporción válida
assert 0 < BASELINE < BASELINE + MDE < 1  ### Comprueba que las proporciones estén entre 0 y 1

# Verificar que el umbral de negocio sea positivo y menor que el MDE
assert 0 < UMBRAL_NEGOCIO < MDE  ### Mantiene diferenciados el criterio estadístico y el criterio empresarial

# ============================================================
# CÁLCULO DEL TAMAÑO DE MUESTRA PARA CONVERSIÓN
# ============================================================

# Crear el objeto para calcular potencia y tamaño de muestra
analysis = NormalIndPower()

# Calcular el tamaño del efecto entre la conversión base y la conversión esperada
es = proportion_effectsize(BASELINE + MDE, BASELINE)  ### Compara 11.8 % frente a 13.3 %

# Calcular la muestra necesaria para detectar el MDE con la potencia establecida
n_conversion = int(np.ceil(analysis.solve_power(
    effect_size=es, alpha=ALPHA, power=POTENCIA,
    ratio=1, alternative="larger")))  ### Prueba unilateral para detectar un aumento de B sobre A

# ============================================================
# CÁLCULO DEL TAMAÑO DE MUESTRA PARA EL GUARDRAIL
# ============================================================

# Obtener los valores críticos de la distribución normal
z_alpha = norm.ppf(1 - ALPHA)  ### Valor crítico asociado al nivel de significancia
z_power = norm.ppf(POTENCIA)  ### Valor asociado a la potencia estadística

# Calcular la muestra aproximada necesaria para evaluar el margen del guardrail
n_guardrail = int(np.ceil(
    2 * BASE_GUARDRAIL * (1 - BASE_GUARDRAIL)
    * (z_alpha + z_power)**2 / MARGEN_GUARDRAIL**2))  ### Se calcula para evaluar el margen de +0.5 pp del guardrail

# ============================================================
# DEFINICIÓN DE LA MUESTRA PLANIFICADA
# ============================================================

# Seleccionar el mayor tamaño de muestra entre conversión y guardrail
N_POR_GRUPO = max(n_conversion, n_guardrail)  ### Se toma el mayor para cubrir ambos objetivos de diseño

# Mostrar los tamaños de muestra calculados
print("Muestra por grupo para conversión:", n_conversion)
print("Muestra aproximada por grupo para guardrail:", n_guardrail)
print("Muestra planificada por grupo:", N_POR_GRUPO)

# Indicar que el análisis debe realizarse una sola vez al completar el experimento
print("Analizar una sola vez al completar muestra y seguimiento.")  ### Evita el peeking antes de completar muestra y seguimiento
```

### **Tamaño de muestra planificado**

Se necesitan **6027 usuarios por grupo** para detectar el MDE de **1.5 puntos porcentuales** en la conversión. Para el guardrail se requieren **8743 usuarios por grupo**, por lo que se toma este valor como muestra final.

En total, el experimento necesitará **17 486 usuarios**, distribuidos entre A y B. Además, el análisis debe realizarse **una sola vez después de completar la muestra y el seguimiento de 7 días**, evitando revisar resultados anticipadamente.

### Paso 3. Generar y validar el conjunto académico
El simulador asigna aleatoriamente el mismo número de usuarios a A y B. Genera una fila por usuario y completa siete días de observación para todos. Supone conversiones de 11,8 % y 13,8 % e incidencias de 1,8 % en ambos grupos. Son parámetros del generador, no resultados que deban recuperarse exactamente. Las respuestas se simulan independientes para simplificar el ejercicio; una empresa real puede presentar dependencias y estacionalidad no representadas aquí.


```python
# ============================================================
# GENERACIÓN DEL CONJUNTO DE DATOS SIMULADO
# ============================================================

# Crear el generador de números aleatorios utilizando la semilla definida
rng = np.random.default_rng(SEMILLA)  ### Permite obtener resultados reproducibles en la simulación

# Calcular el número total de usuarios del experimento
total = 2 * N_POR_GRUPO  ### Se consideran dos grupos: A_Control y B_Tratamiento

# Asignar aleatoriamente el mismo número de usuarios a cada grupo
grupos = rng.permutation(np.repeat(["A_Control", "B_Tratamiento"], N_POR_GRUPO))  ### La permutación evita un orden fijo de asignación

# Definir la probabilidad de compra según el grupo asignado
prob_compra = np.where(grupos == "A_Control", 0.118, 0.138)  ### A tiene 11.8 % y B 13.8 % como parámetros del simulador

# ============================================================
# CONSTRUCCIÓN DE LA TABLA DE USUARIOS
# ============================================================

# Crear una fila por cada usuario con sus variables del experimento
df = pd.DataFrame({

    "usuario_id": np.arange(1, total + 1),  ### Generar un identificador único para cada usuario

    "grupo": grupos,  ### Registrar la variante asignada a cada usuario

    "dia_asignacion": rng.integers(1, DIAS_RECLUTAMIENTO + 1, total),  ### Asignar aleatoriamente el día de ingreso al experimento

    "dias_observados": np.full(total, VENTANA_DIAS),  ### Completar siete días de observación para todos los usuarios

    "compra_7d": rng.binomial(1, prob_compra),  ### Simular si cada usuario realizó una compra dentro de los siete días

    "incidencia_pago_7d": rng.binomial(1, BASE_GUARDRAIL, total)  ### Simular la incidencia de pago utilizando el baseline de 1.8 %
})

# ============================================================
# VALIDACIÓN INICIAL DEL CONJUNTO SIMULADO
# ============================================================

# Mostrar información básica del conjunto de datos generado
print("DATOS SIMULADOS — NexaCommerce")

# Mostrar la cantidad de filas y columnas del conjunto de datos
print("Dimensión:", df.shape)  ### Permite verificar que se generó una fila por usuario y las variables esperadas

# Mostrar las primeras filas para revisar la estructura de los datos
display(df.head())
```

### **Validación del conjunto de datos simulado**

El conjunto generado contiene **17 486 usuarios y 6 variables**, lo que coincide con la muestra planificada de **8743 usuarios por grupo**. Cada usuario tiene un identificador, grupo asignado, día de asignación, siete días de observación, resultado de compra e incidencia de pago.

La tabla muestra que los usuarios fueron asignados a **A_Control o B_Tratamiento** y que todos cuentan con los **7 días de observación**, cumpliendo las condiciones establecidas para el experimento.



```python
# ============================================================
# VALIDACIÓN DE LA CALIDAD Y ESTRUCTURA DE LOS DATOS
# ============================================================

# Definir las columnas indispensables para el análisis
requeridas = {"usuario_id", "grupo", "dias_observados",
              "compra_7d", "incidencia_pago_7d"}

# Verificar que todas las columnas necesarias estén presentes
assert requeridas.issubset(df.columns), "Faltan columnas"

# Verificar que no existan valores ausentes en las variables principales
assert not df[list(requeridas)].isna().any().any(), "Hay datos ausentes"  ### Cada usuario debe tener una observación completa

# Verificar que cada usuario aparezca una sola vez
assert df["usuario_id"].is_unique, "Usuarios repetidos o contaminados"  ### Evita que un usuario aparezca más de una vez

# Verificar que únicamente existan los dos grupos definidos
assert set(df["grupo"]) == {"A_Control", "B_Tratamiento"}

# Verificar que todos los usuarios tengan completa la ventana de seguimiento
assert (df["dias_observados"] == VENTANA_DIAS).all()  ### Todos deben tener los 7 días de observación

# Verificar que las variables de respuesta sean binarias
for col in ["compra_7d", "incidencia_pago_7d"]:
    assert df[col].isin([0, 1]).all(), f"{col} no es binaria"  ### 1 indica que ocurrió el evento y 0 que no ocurrió

# Crear una copia del conjunto validado para continuar con el análisis
ab = df.copy()

# Calcular la cantidad de usuarios asignados a cada grupo
tam_grupos = ab.groupby("grupo").size()

# Verificar que ambos grupos tengan exactamente el tamaño planificado
assert (tam_grupos == N_POR_GRUPO).all()  ### Cada grupo debe contener 8743 usuarios

# Mostrar el número de usuarios por grupo
display(tam_grupos.to_frame("usuarios"))

# Confirmar que las validaciones fueron superadas
print("Calidad estructural y seguimiento completos.")
```

### **Validación de los grupos experimentales**

La validación confirma que ambos grupos tienen exactamente **8743 usuarios**, cumpliendo con el tamaño de muestra planificado. El mensaje **“Calidad estructural y seguimiento completos”** indica que las verificaciones realizadas sobre la estructura de los datos y el seguimiento fueron superadas correctamente.

Por lo tanto, el conjunto de datos queda preparado para continuar con el **análisis de la métrica primaria y del guardrail**.

## Actividad 5 — Análisis estadístico de la comparación A/B

**Propósito:** calcular el efecto observado y evaluar la hipótesis mediante una prueba para dos proporciones.

### Paso 6. Obtener tasas y tamaño del efecto observado

Antes de ejecutar, recuerde:

- Diferencia absoluta: $p_B-p_A$.
- Lift relativo: $(p_B-p_A)/p_A$.
- La diferencia se expresa en puntos porcentuales; el lift expresa un cambio relativo respecto de A.


```python
# ============================================================
# CÁLCULO DE TASAS Y EFECTO OBSERVADO
# ============================================================

# Calcular compras, tamaño de muestra y tasa de conversión por grupo
tabla = ab.groupby("grupo")["compra_7d"].agg(["sum", "count", "mean"])

# Renombrar las columnas para facilitar su interpretación
tabla.columns = ["compras", "n", "tasa_conversion"]

# Obtener el número de compras y usuarios del grupo de control
x_A, n_A = int(tabla.loc["A_Control", "compras"]), int(tabla.loc["A_Control", "n"])  ### x_A representa compras y n_A el total de usuarios de A

# Obtener el número de compras y usuarios del grupo de tratamiento
x_B, n_B = int(tabla.loc["B_Tratamiento", "compras"]), int(tabla.loc["B_Tratamiento", "n"])  ### x_B representa compras y n_B el total de usuarios de B

# Calcular las tasas de conversión de ambos grupos
p_A, p_B = x_A / n_A, x_B / n_B  ### Cada tasa corresponde a compras divididas entre todos los usuarios del grupo

# Calcular la diferencia absoluta de conversión entre B y A
efecto = p_B - p_A  ### Representa el cambio observado en puntos porcentuales

# Calcular el lift relativo respecto al grupo A
lift_rel = efecto / p_A if p_A > 0 else np.nan  ### Mide el cambio porcentual de B respecto a la conversión de A

# Mostrar la tabla de resultados por grupo
display(tabla)

# Mostrar la diferencia absoluta entre las conversiones
print(f"Diferencia B−A: {efecto*100:.3f} pp")

# Mostrar el cambio relativo de B respecto a A
print(f"Lift relativo: {lift_rel:.2%}")
```

### **Tasas de conversión y efecto observado**

El grupo **A_Control** obtuvo una tasa de conversión de **11.392 %**, mientras que **B_Tratamiento** alcanzó **14.137 %**. La diferencia observada entre ambos grupos es de **+2.745 puntos porcentuales**.

El **lift relativo de 24.10 %** indica que la conversión de B fue aproximadamente 24.10 % mayor que la de A en términos relativos. Estos resultados describen el efecto observado en los datos; todavía es necesario realizar la prueba estadística para determinar si existe evidencia suficiente de una diferencia entre los grupos.

### **Pausa de análisis 4**

> La tasa de B es **14.137 %** y la tasa de A es **11.392 %**. La diferencia observada es **2.745 puntos porcentuales**, equivalente a un lift relativo de **24.10 %**. Estos valores describen la muestra, pero todavía no permiten afirmar que **la diferencia observada sea estadísticamente significativa**.


### Paso 7. Contraste e intervalos coherentes
Usamos la aproximación normal de Wald con error estándar no combinado. Se comprueba que haya al menos diez compras y diez no compras en cada grupo; esta regla práctica no sustituye la revisión del diseño.

Para la prueba unilateral frente a cero, p < 0,05 equivale a límite inferior unilateral del 95 % > 0 con el mismo método. El IC bilateral del 95 % se presenta aparte. La función también permite contrastar un umbral distinto de cero. El p-valor no es la probabilidad de que H0 sea verdadera.


```python
# ============================================================
# CONTRASTE DE WALD E INTERVALOS DE CONFIANZA
# ============================================================
def diferencia_wald(x_b, n_b, x_a, n_a, umbral=0.0):

    # Verificación de frecuencias mínimas
    if min(x_b, n_b-x_b, x_a, n_a-x_a) < 10:
        raise ValueError("Frecuencias pequeñas: revisar método antes de continuar")

    pb, pa = x_b/n_b, x_a/n_a  ### Calcula la tasa de conversión observada en cada grupo.

    d = pb-pa  ### Calcula la diferencia absoluta entre B y A.

    se = np.sqrt(pb*(1-pb)/n_b + pa*(1-pa)/n_a)  ### Obtiene el error estándar no combinado de la diferencia.

    z = (d-umbral)/se  ### Estandariza la diferencia respecto al umbral que se desea contrastar.

    return {"d": d, "se": se, "z": z,
            "p_mayor": norm.sf(z),  ### Calcula el p-valor para una alternativa de que B sea mayor que A.
            "p_menor": norm.cdf(z),  ### Calcula el p-valor para una alternativa de que B sea menor que A.
            "li_uni95": d-norm.ppf(1-ALPHA)*se,  ### Calcula el límite inferior unilateral del 95 %.
            "ls_uni95": d+norm.ppf(1-ALPHA)*se,  ### Calcula el límite superior unilateral del 95 %.
            "ic95_low": d-norm.ppf(1-ALPHA/2)*se,  ### Calcula el límite inferior del IC bilateral del 95 %.
            "ic95_high": d+norm.ppf(1-ALPHA/2)*se}  ### Calcula el límite superior del IC bilateral del 95 %.


# ============================================================
# CONTRASTES FRENTE A CERO Y AL UMBRAL EMPRESARIAL
# ============================================================
conv = diferencia_wald(x_B, n_B, x_A, n_A)  ### Evalúa si la diferencia observada supera cero.

negocio = diferencia_wald(x_B, n_B, x_A, n_A, UMBRAL_NEGOCIO)  ### Evalúa si la diferencia supera el umbral empresarial de +1.0 pp.


# ============================================================
# PRESENTACIÓN DE LOS RESULTADOS DEL CONTRASTE
# ============================================================
print(f"z frente a cero: {conv['z']:.4f}")  ### Muestra el estadístico z del contraste frente a cero.

print(f"p unilateral frente a cero: {conv['p_mayor']:.6g}")  ### Muestra el p-valor para determinar si B supera a A.

print(f"Límite inferior unilateral 95%: {conv['li_uni95']*100:.3f} pp")  ### Muestra el límite inferior unilateral en puntos porcentuales.

print(f"IC bilateral 95%: [{conv['ic95_low']*100:.3f}, {conv['ic95_high']*100:.3f}] pp")  ### Presenta el intervalo bilateral para describir la incertidumbre del efecto.

print(f"p unilateral frente a +1,0 pp: {negocio['p_mayor']:.6g}")  ### Muestra el p-valor al contrastar directamente el umbral empresarial.
```

### **Evidencia estadística y superación del umbral empresarial**

El estadístico **z = 5.4436** y el **p-valor = 2.61091e-08** muestran evidencia estadística de que la conversión de **B es mayor que la de A**, ya que el p-valor es menor que **α = 0.05**.

El **límite inferior unilateral del 95 % es +1.916 pp**, por lo que la mejora mínima compatible con este intervalo se mantiene por encima de **0 pp**. Además, el **IC bilateral del 95 % es [1.757, 3.733] pp**, indicando un rango plausible para la diferencia de conversión B−A.

Al contrastar directamente el **umbral empresarial de +1.0 pp**, se obtiene un **p-valor = 0.000269553**, también menor que 0.05. Esto indica evidencia estadística de que la mejora observada **supera el umbral empresarial establecido**.

## Actividad 6 — Del resultado estadístico a la decisión
### Paso 8. Evaluar el guardrail con incertidumbre
El guardrail es la proporción de usuarios con alguna incidencia de pago en siete días. El margen de deterioro admisible es +0,5 pp. La no inferioridad exige que el límite superior unilateral del 95 % esté por debajo de ese margen. No basta obtener p > 0,05 en una prueba de igualdad.


```python
# ============================================================
# ANÁLISIS DEL GUARDRAIL
# ============================================================

# Obtener las incidencias de cada grupo
guardrail = ab.groupby("grupo")["incidencia_pago_7d"].agg(["sum", "count", "mean"])
guardrail.columns = ["incidencias", "n", "tasa_incidencia"]  ### Resume las incidencias y calcula la tasa de incidencia por grupo.

## Obtener las incidencias de cada grupo
g_A = int(guardrail.loc["A_Control", "incidencias"])  ### Extrae las incidencias observadas en A.
g_B = int(guardrail.loc["B_Tratamiento", "incidencias"])  ### Extrae las incidencias observadas en B.

## Contrastar el guardrail frente al margen permitido
guard = diferencia_wald(g_B, n_B, g_A, n_A, MARGEN_GUARDRAIL)  ### Evalúa el deterioro frente al margen permitido de +0.5 pp.

## Mostrar los resultados
display(guardrail)  ### Presenta las incidencias y tasas observadas en cada grupo.
print(f"Cambio del guardrail B−A: {guard['d']*100:.3f} pp")  ### Muestra el cambio observado entre B y A.
print(f"Límite superior unilateral 95%: {guard['ls_uni95']*100:.3f} pp")  ### Muestra el límite superior para evaluar la no inferioridad.
print(f"Margen permitido: {MARGEN_GUARDRAIL*100:.1f} pp")  ### Muestra el deterioro máximo permitido.
print(f"p de no inferioridad: {guard['p_menor']:.6g}")  ### Muestra el p-valor del contraste de no inferioridad.
```

### **Evaluación del guardrail frente al margen permitido**

La tasa de incidencia fue de **1.590 % en A_Control** y **2.013 % en B_Tratamiento**, generando un aumento observado de **+0.423 puntos porcentuales**.

El **p de no inferioridad = 0.351286** tampoco permite demostrar que el deterioro se encuentre dentro del margen permitido. Por ello, el guardrail **no queda protegido estadísticamente**, aunque el cambio observado puntual sea menor que +0.5 pp.
