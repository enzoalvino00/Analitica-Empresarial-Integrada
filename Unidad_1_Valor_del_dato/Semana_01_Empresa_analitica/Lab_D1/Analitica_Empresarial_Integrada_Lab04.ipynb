### **1. Analítica diagnóstica**

**Definición.** Es el nivel de análisis que toma una variación ya medida y la descompone en los factores medibles con los que se relaciona, de modo que se pueda señalar *dónde* y *en qué componente* se originó el movimiento. No inventa una explicación: reparte una cifra observada entre partes que suman exactamente esa cifra.

**Formalmente.** Si un resultado $R$ pasó de $R_0$ a $R_1$, la analítica diagnóstica busca una partición

$$\Delta R = R_1 - R_0 = \sum_{j=1}^{J} E_j$$

donde cada $E_j$ es un efecto atribuible a un factor identificado y la igualdad se cumple de manera **exacta**, no aproximada. Esa exigencia de reconciliación es lo que distingue un diagnóstico de una narrativa: si los efectos no suman la variación observada, existe un factor no modelado y el reparto no es defendible.

**Respuesta a la pregunta de apoyo — ¿qué la distingue de la descriptiva y qué sigue sin poder afirmar?** La descriptiva responde *qué* cambió, *cuánto*, *dónde* y *cuándo*; la diagnóstica responde *con qué factores medibles se relaciona ese cambio*. Lo que la diagnóstica **sigue sin poder afirmar** es que esos factores *hayan producido* el cambio. Una descomposición es una identidad contable, no una relación causal: demuestra que la variación se reparte de cierta manera, no que si se hubiera intervenido sobre uno de los componentes el resultado habría sido distinto. Para afirmar eso se requiere un diseño experimental (asignación aleatoria) o cuasi-experimental (diferencias en diferencias, variables instrumentales, discontinuidad en la regresión), que este conjunto de datos no provee.

### 2. Driver de un resultado

**Definición.** Es un factor que, además de moverse junto con el resultado, forma parte de la aritmética que lo produce: al cambiar el driver, el resultado cambia por construcción, no por coincidencia. En un árbol de drivers cada nivel es una descomposición multiplicativa o aditiva del nivel superior, de manera que el resultado queda expresado como una función explícita de sus componentes.

**Formalmente.** Si el ingreso se escribe como

$$R = N_{\text{clientes}} \times f_{\text{frecuencia}} \times \bar{t}_{\text{ticket}}$$

entonces $N$, $f$ y $\bar{t}$ son drivers de $R$, porque la identidad se cumple por definición. En cambio, «la temperatura del mes» puede correlacionar con $R$ sin ser driver: no aparece en la identidad y su relación con $R$ podría desaparecer al controlar por otra variable.

**Respuesta a la pregunta de apoyo — ¿cómo se comprueba que una variable es driver y no una simple correlación?** Mediante tres comprobaciones sucesivas: (i) **aritmética**, verificando que la variable aparece en la identidad que define el resultado y que la descomposición reconcilia exactamente; (ii) **de estabilidad**, comprobando que la relación se sostiene al segmentar la población en subgrupos (si desaparece dentro de cada segmento, la asociación agregada era un artefacto de composición, fenómeno conocido como paradoja de Simpson); (iii) **de intervención**, que es la única concluyente y exige mover deliberadamente la variable y observar el resultado. Sin la tercera, lo correcto es hablar de *driver contable* y no de *causa*.

### 3. Efecto volumen

**Definición.** Es la parte de la variación del ingreso que se explica únicamente porque se vendieron más o menos unidades en total, manteniendo congelados tanto la composición del catálogo como los precios del periodo base.

**Formalmente.**

$$E_{\text{volumen}} = (Q_1 - Q_0)\cdot \bar{p}_0, \qquad \bar{p}_0 = \sum_{i} s_{0i}\, p_{0i}, \qquad s_{0i} = \frac{q_{0i}}{Q_0}$$

donde $Q_t=\sum_i q_{ti}$ es el volumen total del periodo $t$, $s_{0i}$ la participación del producto $i$ en el volumen del periodo base y $p_{0i}$ su precio efectivo en ese periodo. El término $\bar{p}_0$ es el **precio medio ponderado del periodo base**: el importe que, en promedio, aportaba cada unidad vendida antes de que cambiara nada.

La lectura es directa: el efecto volumen valoriza el cambio de unidades a los precios y a la mezcla de antes. Responde a la pregunta «si lo único que hubiera cambiado fuese la cantidad total vendida, ¿cuánto habría variado el ingreso?».

**Respuesta a la pregunta de apoyo — si el precio no cambia y los ingresos caen, ¿qué indica el efecto volumen?** Indica que la caída proviene de haber vendido menos unidades, y su magnitud cuantifica exactamente cuánto dinero costó esa pérdida de unidades valorizada al precio medio anterior. Si además la mezcla no cambió, el efecto volumen absorbería toda la variación y la lectura sería inequívoca: el problema es de demanda o de disponibilidad, no de política comercial de precios.

---

### 4. Efecto mezcla

**Definición.** Es la parte de la variación del ingreso que se explica porque cambió la **composición** de lo vendido —qué peso tiene cada producto dentro del total de unidades—, manteniendo congelados el volumen total del periodo final y los precios del periodo base. Es el efecto más sutil de los tres y el que con más frecuencia se confunde con el efecto precio.

**Formalmente.**

$$E_{\text{mezcla}} = Q_1 \cdot \sum_i (s_{1i} - s_{0i})\, p_{0i} \;=\; Q_1\left(\bar{p}_0^{\,s_1} - \bar{p}_0\right), \qquad \bar{p}_0^{\,s_1} = \sum_i s_{1i}\, p_{0i}$$

El término $\bar{p}_0^{\,s_1}$ es un **precio medio contrafactual**: el precio medio que se habría observado si se hubiera vendido con la composición del periodo final pero a los precios del periodo base. La diferencia $\bar{p}_0^{\,s_1} - \bar{p}_0$ aísla, en libras por unidad, cuánto se encareció o abarató el «carrito promedio» por el solo hecho de que la demanda se desplazó hacia otros productos.

**Respuesta a la pregunta de apoyo — ¿puede aumentar el ingreso sin vender más unidades ni subir precios?** Sí, y ese es precisamente el efecto mezcla. Si la empresa vende el mismo número total de unidades, con la misma lista de precios, pero una proporción mayor de esas unidades corresponde a productos caros, el ingreso sube sin que haya habido ni más volumen ni un aumento de precios. El caso simétrico —desplazamiento hacia productos baratos— produce una caída de ingreso que el gerente puede leer erróneamente como «bajamos los precios», cuando ningún precio se movió.

---

### 5. Efecto precio

**Definición.** Es la parte de la variación del ingreso que se explica porque cada producto se vendió a un precio efectivo distinto, manteniendo congeladas tanto la cantidad total como la composición del periodo final.

**Formalmente.**

$$E_{\text{precio}} = Q_1 \cdot \sum_i s_{1i}\,(p_{1i} - p_{0i}) \;=\; Q_1\left(\bar{p}_1 - \bar{p}_0^{\,s_1}\right), \qquad p_{ti} = \frac{r_{ti}}{q_{ti}}$$

Obsérvese que $p_{ti}$ **no es el precio de lista** sino el **precio efectivo**: el ingreso del producto dividido entre sus unidades. Esa definición es deliberada, porque absorbe descuentos, promociones, precios diferenciados por cliente y escalas por volumen, es decir, el precio al que la empresa realmente cobró y no al que decía cobrar.

**Respuesta a la pregunta de apoyo — ¿por qué debe calcularse sobre un catálogo común a ambos periodos?** Porque un producto que solo existe en uno de los dos periodos **no tiene precio de comparación**: no hay $p_{0i}$ contra el cual medir su $p_{1i}$, de modo que la diferencia $p_{1i}-p_{0i}$ es indefinida. Incluirlo obligaría a imputar un precio inventado, lo que introduciría en la cifra de precio un movimiento que en realidad es de catálogo. Por eso la descomposición se aplica sobre la intersección de ambos catálogos y las altas y bajas se cuantifican por separado, como dos componentes adicionales del puente.

### 6. Cohorte de clientes

**Definición.** Es el conjunto de clientes que comparten el mes de su **primera compra**, seguido a lo largo del tiempo. La pertenencia a la cohorte es permanente: un cliente que compró por primera vez en marzo de 2010 pertenece a la cohorte de marzo de 2010 para siempre, aunque siga comprando dos años después.

**Formalmente.** Para cada cliente $u$ se define su mes de origen

$$c(u) = \min_{v \in \text{compras}(u)} \text{mes}(v)$$

y su **antigüedad** en cualquier mes $m$ posterior

$$k = 12\,\big(\text{año}(m) - \text{año}(c(u))\big) + \big(\text{mes}(m) - \text{mes}(c(u))\big)$$

de modo que $k=0$ es el mes de la primera compra, $k=1$ el mes siguiente, y así sucesivamente.

**Respuesta a la pregunta de apoyo — ¿por qué la antigüedad se mide desde el mes de la primera compra y no desde enero?** Porque el comportamiento de un cliente depende mucho más del tiempo transcurrido desde que se incorporó que de la fecha del calendario. La caída de actividad más fuerte ocurre en los primeros meses de vida del cliente; si se midiera todo desde enero, un cliente captado en noviembre aparecería con «dos meses de historia» junto a otro captado en enero con doce, y la comparación mezclaría dos cosas distintas: la maduración natural del cliente y el momento del año. Realinear todas las cohortes en $k=0$ las pone en el mismo punto de partida y convierte la comparación en legítima.

---

### 7. Tasa de retención

**Definición.** Es la proporción de los clientes que estrenaron una cohorte que seguía comprando en un mes determinado de antigüedad.

**Formalmente.** Si $A_{c,k}$ es el conjunto de clientes de la cohorte $c$ con al menos una compra en el mes de antigüedad $k$,

$$\text{Ret}_c(k) = \frac{\left|A_{c,k}\right|}{\left|A_{c,0}\right|}\times 100$$

El denominador es siempre el **tamaño inicial de la cohorte**, nunca el del mes anterior. Por construcción $\text{Ret}_c(0)=100\,\%$ para toda cohorte, de modo que lo informativo no es el punto de partida sino la **pendiente** de la curva. Adviértase también que esta definición admite el «regreso»: un cliente que no compró en $k=3$ pero sí en $k=4$ cuenta como activo en $k=4$, razón por la cual la serie no es necesariamente decreciente. En este laboratorio ese fenómeno se observa con claridad: la cohorte de diciembre de 2009 cae a 35.0 % en el mes 1, retrocede a 33.3 % en el mes 2 y **sube** a 42.5 % en el mes 3, porque una parte de los mayoristas compra de manera intermitente y no mensual.

**Respuesta a la pregunta de apoyo — ¿qué significa una celda vacía en la matriz de retención de una cohorte reciente?** Significa **ausencia de información, no abandono**. Una cohorte nacida en octubre de 2011, dentro de un archivo que termina el 9 de diciembre de 2011, sencillamente todavía no ha vivido su mes 6: la celda está vacía porque ese mes aún no ocurrió. Leerla como retención cero es el error clásico del análisis de cohortes, y produce siempre la misma conclusión falsa —«las cohortes recientes se deterioran»— que no es un hallazgo sino un artefacto del truncamiento del periodo de observación.

### 8. Segmentación explicativa

**Definición.** Es la partición de una variación agregada en subgrupos —país, canal, tipo de cliente, categoría— con el propósito de localizar dónde se concentra el movimiento, priorizando por la **contribución absoluta** de cada segmento a la variación total y no por su variación porcentual.

**Formalmente.** Si $\Delta R = \sum_{g} \Delta R_g$ con $g$ recorriendo los segmentos, la prioridad se establece por $|\Delta R_g|$ (libras que el segmento aporta al movimiento) y no por $\Delta R_g / R_{0g}$ (intensidad del deterioro dentro del segmento). Ambas medidas responden preguntas distintas y son complementarias: la primera dice *cuánto dinero está en juego*, la segunda dice *qué tan enfermo está ese mercado*.

**Respuesta a la pregunta de apoyo — ¿por qué el segmento de mayor caída porcentual puede no ser el prioritario?** Porque el porcentaje se calcula sobre la base del propio segmento, y una base pequeña produce porcentajes espectaculares con importes triviales. En este laboratorio, los Emiratos Árabes Unidos caen **−77.82 %**, la mayor caída relativa de todo el conjunto, y sin embargo esa caída equivale a **£ 6 543.67**: apenas un 0.07 % de los ingresos del periodo base. EIRE cae solo **−25.20 %**, pero esa caída vale **£ 88 853.33**, trece veces más dinero. Priorizar por porcentaje llevaría a asignar el equipo comercial al mercado equivocado. La lectura correcta usa las dos dimensiones a la vez, que es precisamente lo que construye el gráfico de dispersión del Ejercicio 5.

# **Marco formal completo del puente precio-volumen-mezcla**

*(Material de apoyo para la explicación en pizarra, bloque de 8:05 a 8:25.)*

## Punto de partida: la identidad del ingreso

El ingreso de un periodo $t$ es la suma, sobre todos los productos, de unidades por precio:

$$R_t = \sum_{i} q_{ti}\, p_{ti}$$

Introduciendo el volumen total $Q_t = \sum_i q_{ti}$ y las participaciones $s_{ti} = q_{ti}/Q_t$ (que suman 1 por construcción), la identidad se reescribe como

$$R_t = Q_t \sum_i s_{ti}\,p_{ti} = Q_t \cdot \bar{p}_t$$

Esta reescritura es el corazón del método: **el ingreso es volumen total multiplicado por precio medio ponderado**, y el precio medio ponderado depende a la vez de los precios individuales y de la composición de la venta. Ahí está el origen de la confusión gerencial que el laboratorio busca disolver: cuando sube $\bar{p}$, el gerente concluye «subimos los precios», cuando en realidad $\bar{p}$ pudo subir sin que ningún $p_i$ se moviera, solo porque cambiaron los $s_i$.

## La descomposición encadenada

La variación total es $\Delta R = Q_1\bar{p}_1 - Q_0\bar{p}_0$. Para repartirla se introduce un **precio medio contrafactual** $\bar{p}_0^{\,s_1} = \sum_i s_{1i}p_{0i}$ —la mezcla nueva valorizada a los precios viejos— y se construyen tres pasos sucesivos, cada uno de los cuales cambia una sola cosa:

| Paso | Qué se mueve | Qué se mantiene fijo | Efecto |
|---|---|---|---|
| 1 | El volumen total, de $Q_0$ a $Q_1$ | Mezcla $s_0$ y precios $p_0$ | $E_{\text{vol}} = (Q_1-Q_0)\,\bar{p}_0$ |
| 2 | La mezcla, de $s_0$ a $s_1$ | Volumen $Q_1$ y precios $p_0$ | $E_{\text{mez}} = Q_1\left(\bar{p}_0^{\,s_1}-\bar{p}_0\right)$ |
| 3 | Los precios, de $p_0$ a $p_1$ | Volumen $Q_1$ y mezcla $s_1$ | $E_{\text{pre}} = Q_1\left(\bar{p}_1-\bar{p}_0^{\,s_1}\right)$ |

## Demostración de que la suma reconcilia exactamente

$$E_{\text{vol}}+E_{\text{mez}}+E_{\text{pre}} = (Q_1-Q_0)\bar{p}_0 + Q_1\bar{p}_0^{\,s_1} - Q_1\bar{p}_0 + Q_1\bar{p}_1 - Q_1\bar{p}_0^{\,s_1}$$

$$= Q_1\bar{p}_0 - Q_0\bar{p}_0 - Q_1\bar{p}_0 + Q_1\bar{p}_1 = Q_1\bar{p}_1 - Q_0\bar{p}_0 = R_1 - R_0 = \Delta R \qquad \blacksquare$$

Los términos $Q_1\bar{p}_0^{\,s_1}$ y $Q_1\bar{p}_0$ se cancelan por pares. **No queda residuo ni término cruzado**, y por eso el `assert` del notebook exige una diferencia menor a un céntimo: cualquier discrepancia superior no es un error de redondeo sino un error de construcción del código.

> **Advertencia metodológica que debe explicitarse ante la clase.** El orden de los pasos no es neutro. Existen otras descomposiciones válidas —la simétrica de Bennet, la de Fisher, la de Törnqvist— que reparten de manera distinta los términos de interacción entre volumen, mezcla y precio. Ninguna es «la verdadera»: todas reconcilian, pero atribuyen cantidades diferentes a cada efecto. Lo que un análisis serio exige no es elegir la correcta, sino **declarar cuál se usó**, porque dos analistas con métodos distintos obtendrán cifras distintas sin que ninguno se haya equivocado. Es exactamente la misma exigencia que se aplica a las reglas de calidad del dato.

## El puente completo cuando el catálogo cambia

La descomposición anterior requiere que un producto exista en ambos periodos. En datos reales eso no ocurre: aquí hay 990 productos **discontinuados** (estaban en P0 y desaparecieron) y 674 productos **nuevos** (aparecieron en P1). Como no tienen precio de comparación, se los cuantifica como bloques enteros y el puente se cierra así:

$$\Delta R_{\text{total}} = \underbrace{E_{\text{vol}}+E_{\text{mez}}+E_{\text{pre}}}_{\text{catálogo común: 3230 productos}} + \underbrace{R_1^{\text{nuevos}}}_{\text{altas}} - \underbrace{R_0^{\text{discontinuados}}}_{\text{bajas}}$$

Ocultar cualquiera de los dos bloques rompe la reconciliación y, lo que es peor, esconde el hallazgo: en este caso las altas y bajas de catálogo **son** la historia principal.

## Verificación numérica del caso, paso a paso

| Magnitud | Valor calculado |

**Lectura de esta tabla en lenguaje de negocio.** Cada unidad vendida del catálogo común pasó de aportar £ 1.6945 a aportar £ 1.8043, un aumento de £ 0.1098 por unidad. Es decir, **la empresa cobró en promedio un poco menos por producto y, aun así, su carrito promedio se encareció, porque vendió cosas distintas.** Quien lea únicamente el ticket promedio (£ 475.95 → £ 508.14, +6.76 %) concluirá lo contrario, y esa es la equivocación que el puente evita.

## ✅ Pregunta 1 — Clasificación previa · RESUELTA

**a) «Los ingresos del segundo año fueron superiores en 2.5 %.» → DESCRIPTIVA.**

Es una medición de un hecho ocurrido: compara dos magnitudes del mismo indicador en dos periodos y reporta la diferencia. No introduce ningún factor explicativo ni afirma nada sobre el mecanismo. Es verificable —la cifra exacta es **+£ 236 086.10, equivalente a +2.51 %**— y es también, como se verá en la Pregunta 5, profundamente engañosa si se la deja sola. **b) «La caída de unidades se concentra en los productos que dejaron de ofrecerse.» → DIAGNÓSTICA.**

Aquí ya no se describe: se **localiza** la variación dentro de una partición del dato. La afirmación relaciona un resultado (la caída de 377 923 unidades, −6.72 %) con un factor medible (el subconjunto de 990 productos discontinuados, que en P0 movían £ 704 357.16 y 496 903 unidades). Es diagnóstica porque señala *dónde* está el movimiento y permite cuantificar cuánto pesa ese componente, pero **no afirma que la desaparición de esos productos haya causado la caída**: podría ser al revés —dejaron de ofrecerse *porque* ya no se vendían— y el dato, por sí solo, no distingue una dirección de la otra.

**c) «Retirar esos productos del catálogo provocó la caída de unidades.» → CAUSAL.**

El verbo «provocó» introduce una afirmación contrafactual: sostiene que, de no haberse retirado los productos, las unidades no habrían caído. Con este conjunto de datos **no puede sostenerse**, por tres razones concretas: (i) existe **causalidad inversa** plausible, ya que un minorista suele descontinuar justamente lo que se vende poco; (ii) existen **factores de confusión** no observados —el conjunto no contiene stock, campañas, precios de la competencia ni decisiones de compra—; y (iii) **no existe grupo de control**: no hay un conjunto comparable de productos mantenidos en catálogo bajo condiciones equivalentes. Sostener (c) exigiría un experimento —retirar productos aleatoriamente en unos mercados y no en otros— o un cuasi-experimento con un contrafactual creíble.

> **Regla de aula que conviene fijar aquí.** El salto indebido no está en la cifra, sino en el verbo. «Se concentra en», «se asocia con», «coincide con» son diagnósticos. «Provocó», «causó», «se debe a», «por culpa de» son causales, y exigen un diseño que este dato no tiene.

> **Nota para el docente.** La huella SHA-256 no es un adorno de seguridad informática. Cumple la misma función que el sello notarial en un contrato: acredita que el archivo analizado es idéntico al publicado, bit por bit. Si un solo valor cambiara —una celda editada a mano, una fila borrada por accidente— la huella sería completamente distinta. Es la manera de que un tercero pueda reproducir el análisis y llegar a la misma cifra.

```python
# Descargamos el conjunto de datos y verificamos que sea exactamente el publicado
# La huella SHA-256 permite comprobar que el archivo no fue alterado

import hashlib, pathlib, urllib.request

# Definimos la URL oficial del conjunto de datos
URL = ("https://raw.githubusercontent.com/Rociosayan/"
       "analitica-empresarial-integrada/main/semana04/datos/"
       "online_retail_II.parquet")

## Guardamos la huella SHA-256 publicada para comparar el archivo descargado
HUELLA_PUBLICADA = "8f64c20d17d38ab02c0dfddda323574e0ee8c34e719df6ea574d16a6e3678e96"

# Definimos el nombre y ubicación del archivo
ARCHIVO = pathlib.Path("online_retail_II.parquet")

# Descargamos el archivo solo si todavía no existe
if not ARCHIVO.exists():
    urllib.request.urlretrieve(URL, ARCHIVO)

# Calculamos la huella SHA-256 del archivo descargado
huella = hashlib.sha256(ARCHIVO.read_bytes()).hexdigest()

# Mostramos el tamaño y la huella para comprobar la trazabilidad
print("Tamaño        :", f"{ARCHIVO.stat().st_size:,} bytes")
print("SHA-256       :", huella)

## Comparamos la huella calculada con la huella publicada
print("¿Coincide?    :",
      "sí" if huella == HUELLA_PUBLICADA else "NO — avise a la docente")
```

## ✅ Pregunta 2 — Decisiones de calidad ·
*De los problemas listados, el de mayor volumen es el de las líneas sin cliente identificado: 243 007 filas, un **22.77 %** del archivo.*

**a) ¿Por qué esas líneas sí pueden usarse para calcular ingresos?**

Porque el ingreso de una línea se calcula con dos campos que están completos: `cantidad × precio`. La ausencia del identificador del cliente no afecta en absoluto esa aritmética. La venta ocurrió, el dinero entró, el producto salió del almacén; lo único que falta es saber **a quién** se le vendió. En un comercio electrónico esas líneas corresponden típicamente a compras de invitado, a pedidos procesados por canales que no capturaban el identificador, o a ventas de mostrador registradas sin cuenta. Excluirlas del cálculo de ingresos equivaldría a sostener que una cuarta parte de las ventas de la empresa no existió, lo cual es falso y produciría un estado de resultados equivocado.

**b) ¿Por qué no pueden usarse para construir cohortes?**

Porque toda la maquinaria de cohortes descansa sobre la posibilidad de **seguir a la misma persona a lo largo del tiempo**. Para asignar a un cliente su mes de origen $c(u)=\min\,\text{mes}(v)$ hay que poder reconocerlo en compras sucesivas; sin identificador, cada línea es un individuo anónimo e irrepetible. Peor aún: si se les asignara un identificador ficticio, cada transacción pasaría a ser un «cliente nuevo que nunca volvió», y la retención calculada se desplomaría artificialmente. Por eso el notebook construye las cohortes solo sobre los **5 852 clientes identificados**, y esa restricción debe declararse en el informe: el análisis de retención describe al subconjunto identificado, no a toda la base de compradores.

**c) ¿Qué distorsión introduciría descartarlas desde el inicio?**

Tres distorsiones simultáneas, todas en el mismo sentido:

1. **Subestimación del nivel de ingresos.** Se perdería el importe facturado en esas 243 007 líneas, cifra nada menor tratándose de casi una de cada cuatro.
2. **Sesgo de composición entre mercados y periodos.** Las compras anónimas no se distribuyen de manera uniforme: dependen del canal de venta, y un canal puede pesar más en un periodo o en un país que en otro. Al eliminarlas, la comparación P0 contra P1 dejaría de ser homogénea y parte de la variación observada sería un artefacto del filtro, no un hecho del negocio.
3. **Pérdida de la trazabilidad de la decisión.** Descartar antes de medir impide saber cuánto se descartó. La regla profesional es la inversa y el notebook la respeta: **medir primero, decidir después y declarar siempre.**

# BLOQUE 2 — Transacciones válidas y comparación temporal

Con el inventario a la vista se construye la tabla de transacciones válidas aplicando cuatro reglas explícitas. Cada regla es, en rigor, **una definición de qué cuenta como venta**.

| Regla | Qué excluye | Justificación económica |
|---|---|---|
| `.unique()` | 34 335 filas duplicadas exactas | Una fila repetida es un artefacto del sistema, no una venta adicional. |
| Factura sin prefijo `C` | 19 494 líneas de cancelación | Una cancelación no es una venta; es su reverso. Incluirla con signo negativo mezclaría dos periodos, porque la devolución de una compra de P0 puede registrarse en P1. |
| `cantidad > 0` y `precio > 0` | Cantidades ≤ 0 y precios ≤ 0 | Sin unidades o sin precio no hay ingreso que medir. Los precios en cero corresponden a muestras, obsequios y ajustes de inventario. |
| Código fuera de `SERVICIOS` | 5 932 líneas de envío, ajustes y comisiones | Los portes (`POST`, `DOT`, `C2`), los ajustes manuales (`M`, `ADJUST`), las comisiones bancarias y las de Amazon no son producto vendido, y su precio unitario no es comparable con el de un producto. |

Después se delimitan los dos periodos. **Se toman doce meses exactos en cada caso, de diciembre a noviembre**, y la razón es decisiva: el conjunto termina el 9 de diciembre de 2011. Incluir ese mes compararía doce meses contra once meses y nueve días, y **fabricaría una caída que no ocurrió**. Quedan fuera 24 753 filas, que el notebook cuantifica y declara en lugar de silenciar.

```python
# Construimos la tabla de transacciones válidas aplicando las reglas de calidad
## Cada filtro define qué registros se consideran ventas para el análisis

validas = (
    crudo
    .unique()  ## Eliminamos filas duplicadas exactas

    .filter(~pl.col("factura").str.starts_with("C"))  ## Excluimos las cancelaciones

    .filter((pl.col("cantidad") > 0) & (pl.col("precio") > 0))  ## Exigimos cantidad y precio positivos

    .filter(~pl.col("codigo").str.contains(SERVICIOS))  ## Excluimos servicios y ajustes

    .with_columns(
        (pl.col("cantidad") * pl.col("precio")).alias("ingreso"),  ## Calculamos el ingreso por línea
        pl.col("fecha").dt.date().alias("dia"),  ## Extraemos la fecha sin la hora
    )
)

# Comparamos las filas originales con las transacciones válidas
print(f"Filas de partida : {crudo.height:>9,}")  ## Total de filas originales
print(f"Filas válidas    : {validas.height:>9,}")  ## Filas que cumplen las reglas de venta

## Calculamos cuántas filas fueron descartadas y qué porcentaje representan
print(f"Descartadas      : {crudo.height - validas.height:>9,} "
      f"({100 * (crudo.height - validas.height) / crudo.height:.2f} %)")
```

### **Transacciones válidas**

- Se partió de **1,067,371 filas** del conjunto original.
- Después de aplicar las reglas de limpieza, quedaron **1,003,340 transacciones válidas**.
- Se descartaron **64,031 filas**, equivalentes al **6.00 %** del total.
- Esto significa que el **94.00 % de los registros** quedó disponible para continuar con el análisis.
- Las transacciones válidas cumplen con las **reglas definidas para considerar una venta**.

```python
# Definimos dos periodos de doce meses exactos para realizar una comparación justa
## Evitamos usar diciembre de 2011 porque sus datos solo llegan hasta el día 9

INICIO_P0, FIN_P0 = pl.date(2009, 12, 1), pl.date(2010, 11, 30)  ## Periodo base: dic-2009 a nov-2010
INICIO_P1, FIN_P1 = pl.date(2010, 12, 1), pl.date(2011, 11, 30)  ## Periodo comparado: dic-2010 a nov-2011

validas = validas.with_columns(
    pl.when((pl.col("dia") >= INICIO_P0) & (pl.col("dia") <= FIN_P0)).then(pl.lit("P0"))
      .when((pl.col("dia") >= INICIO_P1) & (pl.col("dia") <= FIN_P1)).then(pl.lit("P1"))
      .otherwise(None)
      .alias("periodo")
)

# Contamos las filas que no pertenecen a ninguno de los dos periodos
## Estas filas corresponden principalmente a diciembre de 2011, que está incompleto
fuera = validas.filter(pl.col("periodo").is_null()).height

print(f"P0: dic-2009 a nov-2010     P1: dic-2010 a nov-2011")
print(f"Filas fuera de ambos periodos (diciembre de 2011 incompleto): {fuera:,}")  ## Registros excluidos de la comparación

# Conservamos únicamente las filas pertenecientes a P0 o P1
comparables = validas.filter(pl.col("periodo").is_not_null())

## Obtenemos el total de filas que pueden compararse entre ambos periodos
print(f"Filas comparables: {comparables.height:,}")
```

### **Periodos comparables**

- Se definieron dos periodos de **12 meses exactos**: P0 de **dic-2009 a nov-2010** y P1 de **dic-2010 a nov-2011**.
- Se identificaron **24,753 filas fuera de ambos periodos**, correspondientes al tramo incompleto de diciembre de 2011.
- Finalmente, quedaron **978,587 filas comparables** para realizar el análisis entre P0 y P1.
- De esta manera, la comparación se realiza sobre **periodos homogéneos**, evitando que el diciembre de 2011 incompleto distorsione los resultados.

```python
# Realizamos la comparación temporal entre P0 y P1
## Este análisis es descriptivo: muestra qué cambió y cuánto, pero todavía no explica por qué

resumen = (
    comparables
    .group_by("periodo")  ## Agrupamos las transacciones según P0 y P1
    .agg(
        pl.col("ingreso").sum().alias("ingresos"),  ## Calculamos los ingresos totales
        pl.col("cantidad").sum().alias("unidades"),  ## Calculamos las unidades vendidas
        pl.col("factura").n_unique().alias("facturas"),  ## Contamos las facturas únicas
        pl.col("cliente_id").n_unique().alias("clientes"),  ## Contamos los clientes identificados
        pl.col("codigo").n_unique().alias("productos"),  ## Contamos los productos distintos
    )
    .sort("periodo")  ## Ordenamos los resultados de P0 a P1
)

# Separamos los resultados de los dos periodos para calcular sus variaciones
base, final = resumen.row(0, named=True), resumen.row(1, named=True)

print(f"{'Indicador':<12}{'P0':>16}{'P1':>16}{'Variación':>16}{'%':>10}")

for indicador in ["ingresos", "unidades", "facturas", "clientes", "productos"]:
    a, b = base[indicador], final[indicador]
    print(f"{indicador:<12}{a:>16,.0f}{b:>16,.0f}"
          f"{b - a:>16,.0f}{100 * (b - a) / a:>9.2f}%")  ## Calculamos la variación absoluta y porcentual

# Calculamos el ticket promedio dividiendo ingresos entre facturas
## Permite observar cuánto ingreso representa, en promedio, cada factura
ticket_p0 = base["ingresos"] / base["facturas"]
ticket_p1 = final["ingresos"] / final["facturas"]

print(f"\nTicket promedio  P0: £ {ticket_p0:,.2f}   P1: £ {ticket_p1:,.2f}")
```

# **BLOQUE 3 — Puente precio-volumen-mezcla**

Este es el núcleo del laboratorio. El marco formal completo está desarrollado en la sección «Marco formal del puente precio-volumen-mezcla» de este solucionario; aquí se aplica.

Tres precisiones que el docente debe verbalizar antes de ejecutar las celdas:

1. **El precio que se usa es el precio efectivo**, $p_{ti} = r_{ti}/q_{ti}$, no el de lista. Absorbe descuentos, escalas por volumen y precios negociados: es lo que la empresa realmente cobró.
2. **La descomposición solo se aplica al catálogo común**, porque un producto sin contraparte en el otro periodo no tiene precio de comparación.
3. **El orden de los efectos importa.** Cada uno se calcula manteniendo fijo lo que ya se aisló. Otra secuencia produciría otro reparto igualmente válido; por eso se declara la que se usa.

```python
# Calculamos las unidades y el precio efectivo de cada producto en cada periodo
## El precio efectivo representa lo que realmente se cobró por unidad

por_producto = (
    comparables
    .group_by(["periodo", "codigo"])  ## Agrupamos por periodo y producto
    .agg(
        pl.col("cantidad").sum().alias("q"),  ## Sumamos las unidades vendidas
        pl.col("ingreso").sum().alias("r")  ## Sumamos los ingresos generados
    )
    .with_columns(
        (pl.col("r") / pl.col("q")).alias("p")  ## Calculamos el precio efectivo: ingreso / unidades
    )
)

# Separamos la información de P0 y P1 para poder compararlos
p0 = por_producto.filter(pl.col("periodo") == "P0").drop("periodo")  ## Datos del periodo base
p1 = por_producto.filter(pl.col("periodo") == "P1").drop("periodo")  ## Datos del periodo final

# Mostramos cuántos productos existen en cada periodo
print("Productos en P0:", p0.height, "   Productos en P1:", p1.height)
```

### **Productos por periodo**

En P0 se identificaron **4,220 productos**, mientras que en P1 se identificaron **3,904 productos**.

Esto representa una reducción de **316 productos (-7.49 %)** entre ambos periodos. Por lo tanto, el catálogo utilizado en P1 es menor que el de P0.

Para el puente precio-volumen-mezcla, esta diferencia es importante porque la descomposición se realizará sobre el **catálogo común**, es decir, únicamente sobre los productos presentes en ambos periodos.

```python
# Separamos los productos según su presencia en P0 y P1
## El catálogo común permite calcular precio, volumen y mezcla de forma comparable
## Los productos nuevos y discontinuados se cuantifican aparte para mantener la reconciliación

comun = p0.join(p1, on="codigo", suffix="_1")  ## Productos presentes en ambos periodos

discontinuados = p0.join(p1, on="codigo", how="anti")  ## Productos vendidos en P0 pero no en P1

nuevos = p1.join(p0, on="codigo", how="anti")  ## Productos vendidos en P1 pero no en P0

# Mostramos el tamaño y los ingresos asociados a cada grupo
print(f"Catálogo común  : {comun.height:>5} productos")  ## Productos comparables entre P0 y P1

print(f"Discontinuados  : {discontinuados.height:>5} productos   "
      f"£ {discontinuados['r'].sum():>14,.2f}")  ## Ingresos de productos que dejaron de venderse

print(f"Nuevos en P1    : {nuevos.height:>5} productos   "
      f"£ {nuevos['r'].sum():>14,.2f}")  ## Ingresos de productos que aparecieron en P1
```

### **Catálogo común, discontinuados y nuevos**

El **catálogo común** está compuesto por **3,230 productos**, que son los que pueden compararse directamente entre P0 y P1.

En P0 dejaron de venderse **990 productos discontinuados**, que representaban **£704,357.16** en ingresos. Por otro lado, en P1 aparecieron **674 productos nuevos**, que generaron **£2,046,903.92**.

Esto muestra que el cambio del catálogo fue importante: **los productos nuevos generaron muchos más ingresos que los productos discontinuados**. Por eso, estos grupos deben mantenerse separados del cálculo de precio-volumen-mezcla del catálogo común.

```python
# Realizamos la descomposición precio-volumen-mezcla sobre el catálogo común
## Solo usamos productos presentes en P0 y P1 para asegurar una comparación válida

# Calculamos el volumen total de unidades del catálogo común en cada periodo
Q0 = comun["q"].sum()  ## Volumen total del catálogo común en P0
Q1 = comun["q_1"].sum()  ## Volumen total del catálogo común en P1

# Calculamos la participación de cada producto dentro del volumen de su periodo
c = comun.with_columns(
    (pl.col("q") / Q0).alias("s0"),  ## Participación del producto en P0
    (pl.col("q_1") / Q1).alias("s1")  ## Participación del producto en P1
)

# Calculamos el efecto volumen
## Mide cuánto cambia el ingreso por vender más o menos unidades
## manteniendo como referencia los precios y la composición de P0
efecto_volumen = (Q1 - Q0) * (c["s0"] * c["p"]).sum()

# Calculamos el efecto mezcla
## Mide el impacto del cambio en la composición de productos
## manteniendo constantes los precios efectivos de P0
efecto_mezcla = Q1 * ((c["s1"] - c["s0"]) * c["p"]).sum()

# Calculamos el efecto precio
## Mide el impacto del cambio en los precios efectivos de los productos
## utilizando la composición observada en P1
efecto_precio = Q1 * (c["s1"] * (c["p_1"] - c["p"])).sum()

# Calculamos la variación real del ingreso del catálogo común
variacion_comun = c["r_1"].sum() - c["r"].sum()  ## Ingreso P1 menos ingreso P0

# Sumamos los tres efectos para comprobar la reconciliación
suma_efectos = efecto_volumen + efecto_mezcla + efecto_precio  ## Volumen + mezcla + precio

# Mostramos cada componente y comprobamos que la suma coincida con la variación real
print(f"Efecto volumen : £ {efecto_volumen:>14,.2f}")
print(f"Efecto mezcla  : £ {efecto_mezcla:>14,.2f}")
print(f"Efecto precio  : £ {efecto_precio:>14,.2f}")
print(f"{'-' * 34}")
print(f"Suma           : £ {suma_efectos:>14,.2f}")  ## Suma de los tres efectos
print(f"Variación real : £ {variacion_comun:>14,.2f}")  ## Variación real del catálogo común
print(f"Diferencia     : £ {variacion_comun - suma_efectos:>14,.2f}")  ## Debe ser 0 para validar la reconciliación
```

### **Descomposición precio-volumen-mezcla**

El **efecto volumen** fue de **−£1,568,093.05**, mostrando que la reducción de unidades tuvo un impacto negativo importante sobre los ingresos.

El **efecto mezcla** fue de **+£577,550.82**, lo que indica que el cambio en la composición de los productos compensó parcialmente la pérdida de volumen.

El **efecto precio** fue de **−£115,918.44**, por lo que los cambios en los precios efectivos también tuvieron un impacto negativo.

La suma de los tres efectos fue **−£1,106,460.66**, exactamente igual a la **variación real de −£1,106,460.66**. La **diferencia de £0.00 confirma que la descomposición reconcilia correctamente**.

En conclusión, dentro del catálogo común, el **volumen fue el principal efecto negativo**, mientras que la **mezcla compensó una parte importante de la caída**.

```python
# Comprobamos que la descomposición de ingresos sea correcta
## Si los efectos no coinciden con la variación observada, el análisis debe detenerse

assert abs(variacion_comun - suma_efectos) < 0.01, "Los efectos no reconcilian"  ## Verificamos el catálogo común

# Calculamos la variación total de ingresos entre P0 y P1
variacion_total = p1["r"].sum() - p0["r"].sum()  ## Ingreso total de P1 menos ingreso total de P0

# Reconstruimos la variación total incluyendo productos nuevos y discontinuados
reconstruida = variacion_comun + nuevos["r"].sum() - discontinuados["r"].sum()

assert abs(variacion_total - reconstruida) < 0.01, "El puente completo no reconcilia"  ## Verificamos el puente completo

# Mostramos los resultados de la reconciliación
print("Reconciliación verificada.")

print(f"\nVariación total de ingresos : £ {variacion_total:,.2f}")  ## Variación total entre P0 y P1
print(f"  Catálogo común            : £ {variacion_comun:,.2f}")  ## Variación explicada por productos comunes
print(f"    · efecto volumen        : £ {efecto_volumen:,.2f}")  ## Impacto del cambio en unidades
print(f"    · efecto mezcla         : £ {efecto_mezcla:,.2f}")  ## Impacto del cambio en la composición
print(f"    · efecto precio         : £ {efecto_precio:,.2f}")  ## Impacto del cambio en precios efectivos
print(f"  Productos nuevos          : £ {nuevos['r'].sum():,.2f}")  ## Ingresos aportados por productos nuevos
print(f"  Productos discontinuados  : £ {-discontinuados['r'].sum():,.2f}")  ## Ingresos perdidos por productos discontinuados
```

### **Reconciliación completa de los ingresos**

La variación total de ingresos entre P0 y P1 fue de **+£236,086.10**.

El **catálogo común** tuvo una variación negativa de **−£1,106,460.66**, explicada por un efecto volumen de **−£1,568,093.05**, un efecto mezcla positivo de **+£577,550.82** y un efecto precio negativo de **−£115,918.44**.

Los **productos nuevos aportaron +£2,046,903.92**, mientras que los **productos discontinuados representaron −£704,357.16**. Al combinar estos componentes se obtiene exactamente la variación total de **+£236,086.10**.

La **reconciliación fue verificada correctamente**, por lo que el puente completo no presenta diferencias y las cifras pueden utilizarse para el diagnóstico.

# **BLOQUE 4 — Cohortes y curvas de retención**

El puente explicó la variación desde el lado del **producto**. Este bloque la examina desde el lado del **cliente**.

**Advertencia de alcance que debe declararse en el informe.** Las cohortes se construyen únicamente sobre los **5 852 clientes identificados**. El 22.77 % de las líneas sin identificador queda fuera por necesidad metodológica, no por conveniencia: sin identificador no hay a quién seguir. En consecuencia, toda conclusión sobre retención describe al subconjunto identificado y no necesariamente a la totalidad de los compradores.

**Advertencia de truncamiento.** El archivo termina el 9 de diciembre de 2011. Por lo tanto:

- La cohorte de **noviembre de 2011** tiene como mes 1 un diciembre de apenas nueve días: su 14.1 % de retención está **mecánicamente subestimado** y no debe compararse con el de ninguna otra cohorte.
- Las cohortes nacidas después de junio de 2011 **no alcanzan el mes 6**, y sus celdas vacías significan *aún no observado*, nunca *abandonaron*.

**Cómo se lee correctamente una matriz de cohortes.** En diagonal, no en vertical. Cada fila es una cohorte y cada columna una antigüedad; comparar dos celdas solo es legítimo si están en la **misma columna** (misma antigüedad) y ambas corresponden a cohortes que efectivamente vivieron ese mes.

```python
# Construcción de cohortes de clientes
## Una cohorte agrupa clientes según el mes de su primera compra.

clientes = (
    validas
    .filter(pl.col("cliente_id").is_not_null())  ## Conservamos únicamente clientes identificados para poder hacer seguimiento
    .with_columns(
        pl.col("fecha")
        .dt.truncate("1mo")  ## Convertimos cada fecha al inicio de su mes correspondiente
        .dt.date()  ## Convertimos el resultado a formato fecha
        .alias("mes")  ## Guardamos el mes de cada compra en una nueva columna
    )
)

# Identificamos el mes de la primera compra de cada cliente
## Ese primer mes determina la cohorte a la que pertenece el cliente.

clientes = clientes.with_columns(
    pl.col("mes").min().over("cliente_id").alias("cohorte")  ## Encontramos el primer mes de compra de cada cliente
)

# Calculamos la antigüedad de cada cliente desde su primera compra
## La antigüedad nos permite comparar clientes de diferentes cohortes usando los mismos meses de seguimiento.

clientes = clientes.with_columns(
    (
        (pl.col("mes").dt.year() - pl.col("cohorte").dt.year()) * 12  ## Convertimos la diferencia de años a meses
        + (pl.col("mes").dt.month() - pl.col("cohorte").dt.month())  ## Sumamos la diferencia entre los meses
    ).alias("antiguedad")  ## Guardamos la cantidad de meses desde la primera compra
)

# Mostramos la cantidad de clientes identificados
## Solo podemos seguir a clientes que tienen un identificador disponible.

print("Clientes identificados :", f"{clientes['cliente_id'].n_unique():,}")  ## Contamos los clientes únicos identificados

# Mostramos la cantidad de cohortes mensuales
## Cada mes diferente de primera compra representa una cohorte.

print("Cohortes mensuales     :", clientes["cohorte"].n_unique())  ## Contamos las cohortes existentes

# Mostramos algunos registros para comprobar que la construcción sea correcta
## Revisamos cliente, mes de compra, cohorte y antigüedad antes de continuar con el análisis.

print(
    clientes.select(
        "cliente_id", "mes", "cohorte", "antiguedad"  ## Seleccionamos las columnas necesarias para validar el resultado
    ).head(5)  ## Mostramos los primeros 5 registros
)
```

### **Construcción de las cohortes de clientes**

El análisis identificó **5,852 clientes** con identificador disponible y **25 cohortes mensuales**. Esto significa que podemos realizar seguimiento de los clientes identificados según el mes en que realizaron su primera compra.

Los registros de ejemplo confirman que la **cohorte corresponde al primer mes de compra**. Por ejemplo, el cliente **15794** pertenece a la cohorte de diciembre de 2009 y tiene una antigüedad de **0 meses** en ese periodo. El cliente **17841** también pertenece a diciembre de 2009, pero realizó una compra en marzo de 2010, por lo que presenta una antigüedad de **3 meses**.

También se observa que el cliente **16904** pertenece a la cohorte de octubre de 2010 y aparece en agosto de 2011 con una antigüedad de **10 meses**. Esto confirma que la antigüedad se calcula desde la primera compra de cada cliente.

En conclusión, la construcción de las cohortes es correcta y ya tenemos la estructura necesaria para calcular la **retención de clientes por cohorte y por mes de antigüedad**.

```python
# Calculamos la cantidad de clientes activos en cada cohorte y mes de antigüedad
## Un cliente activo es aquel que realizó al menos una compra durante ese mes.
## Contamos clientes únicos para evitar que un mismo cliente sea contado varias veces.

activos = (
    clientes
    .group_by(["cohorte", "antiguedad"])  ## Agrupamos por cohorte y por los meses transcurridos desde la primera compra
    .agg(
        pl.col("cliente_id").n_unique().alias("activos")  ## Contamos cuántos clientes únicos siguieron comprando
    )
)

# Calculamos el tamaño inicial de cada cohorte
## La antigüedad 0 representa el mes en que los clientes realizaron su primera compra.
## Esta cantidad será la base que utilizaremos para calcular la tasa de retención.

tamano_inicial = (
    activos
    .filter(pl.col("antiguedad") == 0)  ## Nos quedamos únicamente con los clientes del mes inicial
    .select(
        "cohorte",
        pl.col("activos").alias("inicial")  ## Guardamos el número inicial de clientes de cada cohorte
    )
)

# Calculamos la tasa de retención de cada cohorte
## La retención indica qué porcentaje de los clientes iniciales continuó comprando en cada mes.
## La fórmula es: clientes activos del mes / clientes iniciales × 100.

retencion = (
    activos
    .join(
        tamano_inicial,
        on="cohorte"
    )  ## Unimos cada cantidad de clientes activos con el tamaño inicial de su cohorte
    .with_columns(
        (100 * pl.col("activos") / pl.col("inicial")).alias("tasa")  ## Calculamos el porcentaje de clientes que permanecen activos
    )
    .sort(
        ["cohorte", "antiguedad"]
    )  ## Ordenamos las cohortes y sus meses para facilitar la lectura
)

# Construimos la matriz de retención
## Cada fila representa una cohorte y cada columna representa un mes de antigüedad.
## Esto permite comparar cohortes que tienen la misma antigüedad.

matriz = (
    retencion
    .pivot(
        values="tasa",  ## Utilizamos la tasa de retención como valor de las celdas
        index="cohorte",  ## Cada fila corresponde a una cohorte
        on="antiguedad",  ## Cada columna representa un mes desde la primera compra
        aggregate_function="first"  ## Tomamos el primer valor cuando existe una única combinación
    )
    .sort("cohorte")  ## Ordenamos las cohortes cronológicamente
)

# Mostramos los primeros seis meses de antigüedad
## Las columnas 0 a 5 permiten observar el comportamiento desde el mes inicial hasta el quinto mes posterior.
## Las celdas vacías significan que todavía no existe información suficiente para observar ese periodo.

print(
    matriz.select(
        ["cohorte"] + [str(i) for i in range(6)]  ## Seleccionamos la cohorte y los meses de antigüedad 0, 1, 2, 3, 4 y 5
    )
)
```

### **Matriz de retención por cohortes**

La matriz muestra **25 cohortes mensuales**, y todas comienzan con una retención de **100 % en el mes 0**, porque ese es el momento en que los clientes realizan su primera compra. A partir del mes 1 podemos observar qué proporción de esos clientes volvió a comprar.

La cohorte de **diciembre de 2009** presenta uno de los comportamientos más fuertes: alcanza **35.02 % en el mes 1** y **35.96 % en el mes 5**. En cambio, otras cohortes presentan retenciones menores; por ejemplo, la cohorte de **febrero de 2010** llega a **19.73 % en el mes 5**. Esto muestra que la permanencia de los clientes no es igual entre todas las cohortes.

Las cohortes más recientes tienen valores `null` en los meses posteriores porque **todavía no existe suficiente información para observar esos periodos**. Por ejemplo, agosto de 2011 no tiene dato para el mes 5, septiembre no tiene datos desde el mes 3 y noviembre de 2011 solo tiene información hasta el mes 1. Estos valores vacíos **no significan 0 % de retención**, sino que el periodo aún no ha sido observado debido al truncamiento del archivo.

Un caso que debe tratarse con especial cuidado es la cohorte de **noviembre de 2011**, que presenta **14.14 % en el mes 1**. Este valor está mecánicamente subestimado porque el archivo termina el **9 de diciembre de 2011**, por lo que no contamos con un mes completo para observar el comportamiento de esa cohorte.

En conclusión, la matriz permite comparar correctamente la retención **por antigüedad**, no simplemente por posición vertical. Para comparar cohortes debemos observar la **misma columna**, ya que representa el mismo número de meses desde la primera compra. Esto nos permitirá posteriormente calcular y analizar las curvas de retención.

```python
# Seleccionamos las columnas que representan los primeros 13 meses de antigüedad
## Esto permite visualizar la retención desde el mes 0 hasta el mes 12 sin mostrar una tabla demasiado extensa.
columnas = [str(i) for i in range(13)]  ## Creamos los nombres de las columnas correspondientes a los meses 0 a 12

tabla = matriz.select(
    ["cohorte"] + [c for c in columnas if c in matriz.columns]
)  ## Seleccionamos la cohorte y únicamente las columnas que realmente existen en la matriz


# Construimos el mapa de calor de la retención
## Cada fila representa una cohorte y cada columna representa un mes de antigüedad.
## Los colores permiten identificar rápidamente dónde la retención es mayor o menor.

figura = px.imshow(
    tabla.drop("cohorte").to_numpy(),  ## Convertimos las tasas de retención en una matriz numérica para graficarlas
    x=[f"Mes {c}" for c in tabla.columns[1:]],  ## Etiquetamos las columnas como Mes 0, Mes 1, Mes 2, etc.
    y=[str(f) for f in tabla["cohorte"].to_list()],  ## Utilizamos el mes de primera compra como nombre de cada cohorte
    color_continuous_scale="Blues",  ## Utilizamos una escala de colores para representar diferentes niveles de retención
    aspect="auto",  ## Permitimos que el gráfico adapte automáticamente el tamaño de las celdas
    labels={"color": "Retención (%)"},  ## Indicamos que la escala de colores representa el porcentaje de retención
    text_auto=".0f",  ## Mostramos el porcentaje de retención dentro de cada celda sin decimales
)


# Configuramos los títulos y dimensiones del gráfico
## Estos elementos ayudan a interpretar qué representa cada eje de la matriz.

figura.update_layout(
    title="Matriz de retención por cohorte mensual",  ## Título que indica que analizamos la retención de cada cohorte
    xaxis_title="Antigüedad",  ## El eje X muestra cuántos meses han pasado desde la primera compra
    yaxis_title="Cohorte (mes de primera compra)",  ## El eje Y identifica el mes en que cada grupo realizó su primera compra
    height=620  ## Definimos una altura suficiente para visualizar las 25 cohortes
)


# Mostramos la matriz de retención
## El mapa de calor facilita comparar visualmente la permanencia de los clientes entre cohortes.
## Las celdas vacías representan periodos que las cohortes recientes todavía no han alcanzado, no una retención de 0 %.

figura.show()  ## Visualizamos el mapa de calor final
```

### **Matriz de retención por cohorte mensual**

El mapa de calor muestra que todas las cohortes comienzan con **100 % de retención en el Mes 0**, ya que ese es el momento de su primera compra. A partir del Mes 1, la retención disminuye y presenta diferentes comportamientos según la cohorte.

La cohorte de **diciembre de 2009** destaca por mantener una retención relativamente alta. Por ejemplo, alcanza **50 % en el Mes 11** y **38 % en el Mes 12**. Esto indica que una parte importante de los clientes de esta cohorte continuó comprando durante un periodo prolongado.

En cambio, algunas cohortes presentan una caída más fuerte. Por ejemplo, la cohorte de **septiembre de 2010** llega a solo **5 % en el Mes 5** y **3 % en el Mes 9**, mostrando una menor permanencia de sus clientes. Esto evidencia que la retención cambia considerablemente entre cohortes.

También se observa claramente el efecto del **truncamiento del archivo**. Las cohortes más recientes tienen celdas vacías en los meses finales porque todavía no han alcanzado esas antigüedades. Por ejemplo, las cohortes de octubre y noviembre de 2011 no pueden compararse con las cohortes antiguas en meses avanzados. Estas celdas vacías **no representan una retención de 0 %**.

En conclusión, la matriz permite identificar qué cohortes mantienen mejor a sus clientes y cuáles pierden actividad rápidamente. Para realizar una comparación correcta debemos hacerlo **por la misma antigüedad**, es decir, comparar cohortes en la misma columna y no simplemente de forma vertical.

```python
# Seleccionamos únicamente las cohortes que ya tienen al menos seis meses de observación
## Esto permite comparar las curvas hasta el Mes 6 sin incluir cohortes recientes incompletas.

maduras = (
    retencion
    .join(
        retencion
        .group_by("cohorte")
        .agg(
            pl.col("antiguedad").max().alias("tope")  ## Encontramos el último mes de antigüedad disponible para cada cohorte
        ),
        on="cohorte"  ## Unimos este límite con los registros originales de cada cohorte
    )
    .filter(
        (pl.col("tope") >= 6) &  ## Conservamos solo las cohortes que alcanzaron al menos seis meses
        (pl.col("antiguedad") <= 6)  ## Limitamos el análisis hasta el Mes 6 para comparar el mismo periodo
    )
    .with_columns(
        pl.col("cohorte").cast(pl.Utf8)  ## Convertimos la cohorte a texto para utilizarla como categoría en el gráfico
    )
)

# Convertimos los datos a formato compatible con Plotly y construimos las curvas de retención
## Cada línea representa una cohorte y muestra cómo cambia su retención desde el Mes 0 hasta el Mes 6.

figura = px.line(
    maduras.to_pandas(),  ## Convertimos la tabla de Polars a pandas para facilitar su uso con Plotly
    x="antiguedad",  ## El eje X representa los meses transcurridos desde la primera compra
    y="tasa",  ## El eje Y representa el porcentaje de clientes que continúa activo
    color="cohorte",  ## Cada cohorte tendrá una línea diferente para poder compararlas
    markers=True,  ## Colocamos puntos en cada mes para identificar fácilmente los valores
    labels={
        "antiguedad": "Meses desde la primera compra",  ## Nombre descriptivo del eje X
        "tasa": "Retención (%)",  ## Nombre descriptivo del eje Y
        "cohorte": "Cohorte"  ## Nombre que identifica las diferentes líneas
    }
)

# Configuramos la presentación del gráfico
## El título indica que estamos comparando la evolución de la retención entre diferentes cohortes.

figura.update_layout(
    title="Curvas de retención por cohorte",  ## Título principal del gráfico
    height=480  ## Definimos la altura para que las curvas sean fáciles de visualizar
)

# Mostramos las curvas de retención
## Todas las curvas comienzan en 100 % en el Mes 0 por definición.
## Lo importante es observar la pendiente y cómo cambia la retención durante los seis meses.

figura.show()  ## Visualizamos el gráfico final
```

### **Curvas de retención por cohorte**

Las curvas muestran que todas las cohortes comienzan en **100 % de retención en el Mes 0**, porque ese es el momento de su primera compra. La caída más fuerte ocurre entre el **Mes 0 y el Mes 1**, donde la mayoría de las cohortes baja aproximadamente al rango de **15 % a 30 %**.

**c) ¿Por qué no puede concluirse de esta matriz que la calidad del servicio empeoró?**

# **BLOQUE 5 — Segmentación, driver dominante y tablero**

La segmentación localiza **dónde** se concentra el movimiento. Se examinan dos dimensiones: el país del cliente y su condición de nuevo o recurrente.

**El criterio de priorización es la contribución absoluta en libras, no la variación porcentual.** La razón es aritmética: el porcentaje se calcula sobre la base del propio segmento, de modo que una base pequeña produce porcentajes espectaculares con importes irrelevantes. En este conjunto hay 43 mercados; cinco de ellos —Tailandia, Nigeria, Corea, Indias Occidentales y Bermudas— caen **−100 %** porque simplemente dejaron de registrar ventas, y entre los cinco suman £ 6 091.17, menos del 0.07 % de los ingresos del periodo base.

```python
# Segmentación explicativa por país
## Buscamos identificar qué países explican una mayor cantidad de ingresos perdidos o ganados.
## También calculamos el cambio porcentual para saber qué países se deterioraron más respecto a su propio tamaño.

por_pais = (
    comparables
    .group_by(["pais", "periodo"])  ## Agrupamos las ventas por país y por periodo P0 o P1
    .agg(
        pl.col("ingreso").sum()  ## Sumamos los ingresos de cada país en cada periodo
    )
    .pivot(
        values="ingreso",  ## Utilizamos el ingreso como valor que queremos comparar
        index="pais",  ## Cada fila representa un país
        on="periodo",  ## Creamos una columna para P0 y otra para P1
        aggregate_function="first"  ## Tomamos el único valor resultante para cada país y periodo
    )
    .fill_null(0.0)  ## Si un país no tiene ingresos en alguno de los periodos, colocamos 0
    .with_columns(
        (pl.col("P1") - pl.col("P0")).alias("variacion")  ## Calculamos cuánto aumentó o disminuyó el ingreso en libras
    )
    .with_columns(
        pl.when(pl.col("P0") > 0)  ## Verificamos que el ingreso inicial sea mayor que cero antes de calcular el porcentaje
          .then(
              100 * pl.col("variacion") / pl.col("P0")  ## Calculamos la variación porcentual respecto a P0
          )
          .otherwise(None)  ## Si P0 es cero, evitamos una división entre cero
          .alias("variacion_pct")  ## Guardamos el resultado como variación porcentual
    )
    .sort("variacion")  ## Ordenamos desde las mayores caídas hasta los mayores aumentos
)

# Mostramos los países con las mayores caídas de ingresos
## Al ordenar por "variacion", los valores más negativos aparecen primero.
## Estos países son los que más ingresos perdieron en términos absolutos.

print("Mayores caídas en libras:")
print(por_pais.head(6))  ## Mostramos los 6 países con mayor pérdida absoluta


# Mostramos los países con los mayores aumentos de ingresos
## Los valores positivos más altos aparecen al final de la tabla porque está ordenada de menor a mayor.

print("\nMayores aumentos en libras:")
print(por_pais.tail(6))  ## Mostramos los 6 países con mayor aumento absoluto


# La variación absoluta y porcentual deben analizarse juntas
## Una caída de -70 % no necesariamente representa un problema grande si el país tenía pocos ingresos.
## Por eso, primero priorizamos el impacto absoluto en libras y luego usamos el porcentaje como señal complementaria.
```

### **Segmentación de ingresos por país**

Las mayores caídas en términos absolutos se concentran principalmente en **EIRE**, que pasó de **£352,563.30 a £263,709.97**, representando una pérdida de **£88,853.33 (−25.20 %)**.
```python
# Segunda dimensión de segmentación: clientes nuevos frente a clientes recurrentes
## Buscamos identificar si los ingresos de P1 provienen de clientes que ya compraban o de clientes que compraron por primera vez.

clientes_p0 = (
    comparables
    .filter(pl.col("periodo") == "P0")  ## Nos quedamos únicamente con las compras realizadas durante P0
    ["cliente_id"]
    .unique()  ## Eliminamos clientes repetidos para obtener una sola vez cada cliente
    .to_list()  ## Convertimos los identificadores a una lista para poder utilizarlos en la clasificación
)

# Clasificamos a los clientes de P1 según si ya habían comprado durante P0
## Si el cliente aparece en P0, lo consideramos recurrente; si no aparece, lo consideramos nuevo.

tipos = (
    comparables
    .filter(
        (pl.col("periodo") == "P1") &  ## Analizamos únicamente las compras del periodo P1
        pl.col("cliente_id").is_not_null()  ## Excluimos registros sin identificador porque no podemos clasificar al cliente
    )
    .with_columns(
        pl.when(
            pl.col("cliente_id").is_in(clientes_p0)  ## Comprobamos si el cliente ya había comprado durante P0
        )
        .then(
            pl.lit("recurrente")  ## Si estaba en P0, clasificamos al cliente como recurrente
        )
        .otherwise(
            pl.lit("nuevo")  ## Si no estaba en P0, clasificamos al cliente como nuevo
        )
        .alias("tipo")  ## Guardamos la clasificación en una nueva columna llamada tipo
    )
    .group_by("tipo")  ## Agrupamos los resultados entre clientes nuevos y recurrentes
    .agg(
        pl.col("ingreso")
        .sum()
        .round(2)
        .alias("ingresos"),  ## Calculamos los ingresos totales generados por cada tipo de cliente

        pl.col("cliente_id")
        .n_unique()
        .alias("clientes")  ## Contamos cuántos clientes únicos existen en cada grupo
    )
    .with_columns(
        (
            pl.col("ingresos") / pl.col("clientes")
        )
        .round(2)
        .alias("ingreso_por_cliente")  ## Calculamos cuánto ingreso genera, en promedio, cada cliente de cada grupo
    )
)

# Mostramos la comparación entre clientes nuevos y recurrentes
## La tabla permite conocer cuántos clientes hay en cada grupo, cuánto ingreso generan y cuál es su ingreso promedio por cliente.

print(tipos)
```

La diferencia también se observa en el ingreso promedio por cliente. Cada cliente recurrente generó en promedio **£2,502.51**, mientras que cada cliente nuevo generó **£913.70**. Esto significa que, en promedio, un cliente recurrente generó aproximadamente **2.7 veces más ingresos** que un cliente nuevo.

Por lo tanto, los clientes recurrentes tienen un peso mucho mayor en los ingresos de P1. Esto indica que **mantener a los clientes existentes es especialmente importante para el negocio**, ya que perder clientes recurrentes podría tener un impacto considerable en los ingresos.

En conclusión, aunque los clientes nuevos aportan **£1.45 millones** y son importantes para seguir creciendo, los **clientes recurrentes son el segmento de mayor valor económico**, con **£6.78 millones** de ingresos y un ingreso promedio de **£2,502.51 por cliente**.

```python
# Repetimos el análisis de segmentación por país utilizando SQL
## DuckDB permite consultar directamente el DataFrame de Polars sin necesidad de crear una base de datos.
## Esto demuestra que el mismo análisis puede realizarse tanto con Polars como con SQL.

consulta = duckdb.sql("""
    SELECT pais,  -- Seleccionamos el país para comparar los resultados por mercado

           SUM(
               CASE
                   WHEN periodo = 'P0' THEN ingreso
                   ELSE 0
               END
           ) AS p0,  -- Calculamos los ingresos correspondientes al periodo P0

           SUM(
               CASE
                   WHEN periodo = 'P1' THEN ingreso
                   ELSE 0
               END
           ) AS p1,  -- Calculamos los ingresos correspondientes al periodo P1

           SUM(
               CASE
                   WHEN periodo = 'P1' THEN ingreso
                   ELSE -ingreso
               END
           ) AS variacion,  -- Calculamos la diferencia entre los ingresos de P1 y P0

           COUNT(
               DISTINCT cliente_id
           ) AS clientes  -- Contamos los clientes únicos asociados a cada país

    FROM comparables  -- Utilizamos la tabla de transacciones comparables entre P0 y P1

    GROUP BY pais  -- Agrupamos todas las operaciones para obtener un resultado por país

    HAVING SUM(
        CASE
            WHEN periodo = 'P0' THEN ingreso
            ELSE 0
        END
    ) > 20000  -- Consideramos únicamente países que superaron £20,000 de ingresos en P0
               -- Esto evita priorizar mercados muy pequeños con variaciones porcentuales exageradas

    ORDER BY variacion  -- Ordenamos desde las mayores caídas hasta los mayores aumentos

""").pl()  ## Convertimos el resultado de DuckDB nuevamente a un DataFrame de Polars


# Mostramos el resultado de la consulta SQL
## La tabla permite identificar qué países tuvieron mayor impacto absoluto en la variación de ingresos.
## Al ordenar por "variacion", los países con pérdidas aparecen primero y los que crecieron aparecen después.

print(consulta)  ## Mostramos la tabla final obtenida mediante SQL
```

```python
# ===== EJERCICIO 1 =====
# Identificamos y ponderamos el efecto dominante de la descomposición.
## Comparamos Volumen, Mezcla y Precio según el tamaño de su contribución.
## Usamos el valor absoluto porque queremos medir la magnitud del efecto sin importar si es positivo o negativo.

efectos = pl.DataFrame({
    "efecto": ["Volumen", "Mezcla", "Precio"],  ## Creamos una columna con el nombre de cada efecto
    "contribucion": [efecto_volumen, efecto_mezcla, efecto_precio],  ## Guardamos la contribución calculada de cada efecto
}).with_columns(
    pl.col("contribucion").abs().alias("magnitud"),  ## Convertimos cada contribución a valor absoluto para comparar su tamaño
    (100 * pl.col("contribucion") / variacion_total).alias("pct_de_variacion_neta"),  ## Calculamos qué porcentaje representa cada efecto respecto a la variación neta
).sort("magnitud", descending=True)  ## Ordenamos los efectos desde el de mayor hasta el de menor magnitud


# Mostramos la tabla de efectos ordenada por magnitud.
## Esto permite identificar visualmente cuál es el efecto dominante.
## Redondeamos las cifras para que el resultado sea más fácil de leer.

print(efectos.select(
    "efecto",  ## Mostramos el nombre del efecto
    pl.col("contribucion").round(2),  ## Mostramos la contribución con dos decimales
    pl.col("magnitud").round(2),  ## Mostramos la magnitud absoluta con dos decimales
    pl.col("pct_de_variacion_neta").round(1),  ## Mostramos el porcentaje respecto a la variación neta con un decimal
))


# Identificamos el efecto que ocupa la primera posición después de ordenar.
## Como la tabla está ordenada de mayor a menor magnitud, la primera fila corresponde al efecto dominante.

dominante = efectos.row(0, named=True)  ## Extraemos toda la información del efecto dominante
suma_abs = efectos["magnitud"].sum()  ## Sumamos las magnitudes absolutas de Volumen, Mezcla y Precio


# Mostramos los resultados principales de la priorización.
## Comprobamos cuál es el efecto dominante, cuánto aporta y qué relación existe entre las magnitudes y la variación neta.

print(f"\nEfecto dominante        : {dominante['efecto']}")  ## Mostramos el nombre del efecto con mayor magnitud
print(f"Contribucion            : £ {dominante['contribucion']:,.2f}")  ## Mostramos cuánto contribuye el efecto dominante
print(f"Representa              : {dominante['pct_de_variacion_neta']:,.1f} % de la variacion neta")  ## Mostramos su peso respecto a la variación neta

print(f"\nSuma de magnitudes      : £ {suma_abs:,.2f}")  ## Mostramos la suma de los valores absolutos de los tres efectos
print(f"Variacion neta          : £ {variacion_total:,.2f}")  ## Mostramos la variación neta final
print(f"Razon suma/neta         : {suma_abs / abs(variacion_total):.2f} veces")  ## Calculamos cuántas veces la suma de magnitudes supera a la variación neta
print("Se compensan entre si   :", "SI" if suma_abs > abs(variacion_total) else "NO")  ## Verificamos si existen efectos positivos y negativos que se compensan
```

## Ejercicio 2 — Aplicación de la descomposición a un mercado específico

**Se solicita:** identificar el país con mayor caída en libras excluyendo el Reino Unido; delimitar sus transacciones en ambos periodos; replicar la descomposición volumen-mezcla-precio sobre el catálogo común de ese mercado; comprobar la reconciliación.

```python
# ===== EJERCICIO 2 =====
# Identificamos el mercado con la mayor caída absoluta, excluyendo al Reino Unido.
## Ordenamos los países por su variación de ingresos de menor a mayor.
## El primer registro será el país con la caída más grande en libras.

objetivo = (por_pais
            .filter(pl.col("pais") != "United Kingdom")  ## Excluimos Reino Unido porque el ejercicio pide analizar otro mercado
            .sort("variacion")  ## Ordenamos de la mayor caída a la mayor subida
            .row(0, named=True))  ## Seleccionamos el primer país y lo convertimos en un registro

PAIS = objetivo["pais"]  ## Guardamos el nombre del mercado seleccionado

print(f"Mercado seleccionado: {PAIS}")  ## Mostramos el país elegido
print(f"  P0: £ {objetivo['P0']:,.2f}   P1: £ {objetivo['P1']:,.2f}")  ## Mostramos sus ingresos en ambos periodos
print(f"  Variacion: £ {objetivo['variacion']:,.2f} ({objetivo['variacion_pct']:.2f} %)\n")  ## Mostramos la caída absoluta y porcentual


# Filtramos las transacciones del mercado seleccionado.
## Trabajamos únicamente con las operaciones de ese país para replicar la descomposición a nivel de mercado.

sub = comparables.filter(pl.col("pais") == PAIS)  ## Conservamos solo las transacciones correspondientes al país seleccionado


# Calculamos cantidades e ingresos por producto y periodo.
## Agrupamos por periodo y código de producto para obtener las unidades y el ingreso de cada referencia.
## El precio efectivo se calcula como ingreso dividido entre cantidad.

por_producto_pais = (
    sub.group_by(["periodo", "codigo"])
       .agg(
           pl.col("cantidad").sum().alias("q"),  ## Sumamos las unidades vendidas de cada producto
           pl.col("ingreso").sum().alias("r")  ## Sumamos el ingreso generado por cada producto
       )
       .with_columns(
           (pl.col("r") / pl.col("q")).alias("p")  ## Calculamos el precio efectivo promedio de cada producto
       )
)


# Separamos los productos de P0 y P1.
## Necesitamos comparar las mismas referencias entre ambos periodos.

a0 = por_producto_pais.filter(pl.col("periodo") == "P0").drop("periodo")  ## Extraemos los productos vendidos en P0
a1 = por_producto_pais.filter(pl.col("periodo") == "P1").drop("periodo")  ## Extraemos los productos vendidos en P1


# Clasificamos las referencias según su presencia en ambos periodos.
## Los productos comunes permiten realizar la descomposición volumen-mezcla-precio.
## Los discontinuados solo aparecen en P0 y los nuevos solo aparecen en P1.

comun_pais = a0.join(a1, on="codigo", suffix="_1")  ## Conservamos productos presentes tanto en P0 como en P1
disc_pais = a0.join(a1, on="codigo", how="anti")  ## Identificamos productos que estaban en P0 pero desaparecen en P1
nuev_pais = a1.join(a0, on="codigo", how="anti")  ## Identificamos productos que aparecen en P1 pero no estaban en P0


# Calculamos las unidades totales del catálogo común.
## QA0 representa las unidades de los productos comunes en P0.
## QA1 representa las unidades de los mismos productos comunes en P1.

QA0, QA1 = comun_pais["q"].sum(), comun_pais["q_1"].sum()  ## Sumamos las cantidades del catálogo común en ambos periodos


# Calculamos la participación de cada producto dentro del volumen total del catálogo común.
## s0 corresponde a la participación del producto en P0.
## s1 corresponde a la participación del producto en P1.
## Estas participaciones permiten medir el cambio de mezcla entre los productos.

cp = comun_pais.with_columns(
    (pl.col("q") / QA0).alias("s0"),  ## Calculamos la participación del producto en el volumen de P0
    (pl.col("q_1") / QA1).alias("s1")  ## Calculamos la participación del producto en el volumen de P1
)


# Calculamos los tres efectos de la descomposición volumen-mezcla-precio.
## Volumen mide el impacto de vender más o menos unidades manteniendo la estructura inicial.
## Mezcla mide el impacto de cambiar la participación relativa de los productos.
## Precio mide el impacto del cambio en el precio efectivo de los productos.

vol_pais = (QA1 - QA0) * (cp["s0"] * cp["p"]).sum()  ## Calculamos el efecto volumen del catálogo común

mez_pais = QA1 * ((cp["s1"] - cp["s0"]) * cp["p"]).sum()  ## Calculamos el efecto mezcla debido al cambio en la composición de ventas

pre_pais = QA1 * (cp["s1"] * (cp["p_1"] - cp["p"])).sum()  ## Calculamos el efecto precio debido al cambio de precio efectivo


# Calculamos las variaciones que utilizaremos para comprobar la reconciliación.
## La variación común compara únicamente los productos presentes en ambos periodos.
## La variación total incluye además los productos nuevos y discontinuados.

var_comun_pais = cp["r_1"].sum() - cp["r"].sum()  ## Calculamos el cambio de ingresos del catálogo común

var_total_pais = a1["r"].sum() - a0["r"].sum()  ## Calculamos el cambio total de ingresos del país


# Mostramos la composición del catálogo del mercado seleccionado.
## Esto permite conocer cuántos productos son comunes, nuevos y discontinuados.

print(f"Catalogo comun  : {comun_pais.height:>4} productos")  ## Mostramos la cantidad de productos presentes en ambos periodos
print(f"Discontinuados  : {disc_pais.height:>4} productos   £ {disc_pais['r'].sum():>12,.2f}")  ## Mostramos productos que desaparecieron y sus ingresos de P0
print(f"Nuevos en P1    : {nuev_pais.height:>4} productos   £ {nuev_pais['r'].sum():>12,.2f}\n")  ## Mostramos productos nuevos y sus ingresos de P1


# Mostramos los tres efectos calculados para el catálogo común.
## La suma de estos tres efectos debe coincidir exactamente con la variación del catálogo común.

print(f"Efecto volumen  : £ {vol_pais:>12,.2f}")  ## Mostramos cuánto aporta el cambio de volumen
print(f"Efecto mezcla   : £ {mez_pais:>12,.2f}")  ## Mostramos cuánto aporta el cambio de mezcla
print(f"Efecto precio   : £ {pre_pais:>12,.2f}")  ## Mostramos cuánto aporta el cambio de precio
print(f"{'-'*32}")  ## Dibujamos una línea para separar los efectos del resultado de reconciliación
print(f"Suma            : £ {vol_pais + mez_pais + pre_pais:>12,.2f}")  ## Sumamos los tres efectos
print(f"Variacion comun : £ {var_comun_pais:>12,.2f}")  ## Mostramos la variación real del catálogo común


# Verificamos que la descomposición volumen-mezcla-precio reconcilie.
## La diferencia permitida es menor a £0.01 debido a posibles diferencias de precisión decimal.

assert abs(var_comun_pais - (vol_pais + mez_pais + pre_pais)) < 0.01, "No reconcilia"  ## Comprobamos que los tres efectos expliquen exactamente la variación común


# Reconstruimos la variación total del mercado.
## Partimos de la variación del catálogo común y agregamos los productos nuevos y discontinuados.

reconstruida = var_comun_pais + nuev_pais["r"].sum() - disc_pais["r"].sum()  ## Construimos nuevamente la variación total incluyendo altas y bajas de productos

assert abs(var_total_pais - reconstruida) < 0.01, "El puente del pais no reconcilia"  ## Comprobamos que el puente completo del mercado también reconcilie


# Confirmamos que ambas reconciliaciones fueron correctas.
## Si alguno de los assert anteriores falla, Python detendrá la ejecución y mostrará el error.

print("\nReconciliacion verificada.")  ## Confirmamos que los cálculos son consistentes
print(f"Variacion total de {PAIS}: £ {var_total_pais:,.2f}")  ## Mostramos la variación total de ingresos del mercado

# Comparamos el volumen del catálogo común entre P0 y P1.
## Esto permite identificar si la cantidad de unidades vendidas de las referencias comunes aumentó o disminuyó.

print(f"  Unidades del catalogo comun: {QA0:,} -> {QA1:,} ({100*(QA1-QA0)/QA0:.2f} %)")  ## Mostramos las unidades iniciales, finales y su variación porcentual
```

## Ejercicio 3 — Comparación de la retención entre cohortes

**Se solicita:** extraer la retención del sexto mes de antigüedad de cada cohorte; excluir explícitamente las cohortes que no lo alcanzaron; ordenar cronológicamente y comparar; determinar si existe tendencia de mejora o de deterioro.

```python
# ===== EJERCICIO 3 =====
# Comparamos la retención de las cohortes exactamente en el sexto mes de antigüedad.
## Primero identificamos qué cohortes ya tienen información suficiente para llegar al mes 6.
## Las cohortes que todavía no alcanzaron ese periodo se excluyen explícitamente.

CORTE = 6  ## Definimos el sexto mes como el punto de comparación entre cohortes


# Identificamos las cohortes que alcanzaron al menos 6 meses de antigüedad.
## Para cada cohorte buscamos su mayor antigüedad disponible.
## Si la antigüedad máxima es menor que 6, esa cohorte todavía no puede ser evaluada en el mes 6.

maduras = (retencion
           .group_by("cohorte")  ## Agrupamos los registros según el mes de nacimiento de la cohorte
           .agg(pl.col("antiguedad").max().alias("antiguedad_maxima"))  ## Obtenemos la máxima antigüedad disponible para cada cohorte
           .filter(pl.col("antiguedad_maxima") >= CORTE))  ## Conservamos únicamente las cohortes que alcanzaron el sexto mes


# Extraemos únicamente el registro correspondiente al sexto mes de cada cohorte madura.
## Así obtenemos una comparación homogénea: todas las cohortes se evalúan con exactamente 6 meses de antigüedad.

ret6 = (retencion
        .join(maduras, on="cohorte", how="inner")  ## Conservamos solamente las cohortes que alcanzaron el mes 6
        .filter(pl.col("antiguedad") == CORTE)  ## Seleccionamos exclusivamente el sexto mes de antigüedad
        .select(
            "cohorte",  ## Mostramos el mes en que nació cada cohorte
            pl.col("inicial").alias("clientes_iniciales"),  ## Mostramos cuántos clientes tenía inicialmente la cohorte
            pl.col("activos").alias("activos_mes_6"),  ## Mostramos cuántos clientes permanecen activos en el mes 6
            pl.col("tasa").round(1).alias("retencion_mes_6")  ## Mostramos la tasa de retención al sexto mes
        )
        .sort("cohorte"))  ## Ordenamos las cohortes cronológicamente


# Calculamos cuántas cohortes quedaron fuera del análisis.
## Una cohorte excluida no significa que tenga 0% de retención.
## Significa que todavía no existe información suficiente para medir su sexto mes.

excluidas = retencion["cohorte"].n_unique() - ret6.height  ## Restamos las cohortes evaluables del total de cohortes


# Mostramos el tamaño de la muestra utilizada en la comparación.
## Esto permite comprobar que las cohortes recientes fueron excluidas correctamente.

print(f"Cohortes totales                 : {retencion['cohorte'].n_unique()}")  ## Mostramos el número total de cohortes
print(f"Cohortes que alcanzaron el mes {CORTE} : {ret6.height}")  ## Mostramos cuántas cohortes tienen información del sexto mes
print(f"Cohortes excluidas (sin dato)    : {excluidas}\n")  ## Mostramos cuántas cohortes no pudieron ser evaluadas

print(ret6)  ## Mostramos la retención del sexto mes ordenada cronológicamente


# Dividimos las cohortes evaluables en dos grupos para comparar las más antiguas con las más recientes.
## La división permite observar si la retención está mejorando o deteriorándose a través del tiempo.

mitad = ret6.height // 2  ## Calculamos la mitad del número de cohortes disponibles

antigua = ret6.head(mitad)["retencion_mes_6"].mean()  ## Calculamos la retención promedio del sexto mes de las cohortes más antiguas
reciente = ret6.tail(ret6.height - mitad)["retencion_mes_6"].mean()  ## Calculamos la retención promedio del sexto mes de las cohortes más recientes


# Mostramos la comparación entre las cohortes antiguas y recientes.
## Una diferencia negativa indica que las cohortes recientes retienen menos clientes que las antiguas.

print(f"\nPromedio {mitad} cohortes mas antiguas : {antigua:.2f} %")  ## Mostramos el promedio de retención de las cohortes más antiguas
print(f"Promedio {ret6.height-mitad} cohortes mas recientes: {reciente:.2f} %")  ## Mostramos el promedio de retención de las cohortes más recientes
print(f"Diferencia                        : {reciente - antigua:+.2f} puntos porcentuales")  ## Calculamos el cambio entre ambos grupos


# Determinamos automáticamente si existe mejora o deterioro.
## Si la retención reciente es menor que la antigua, clasificamos el comportamiento como deterioro.

print("Tendencia                         :",
      "DETERIORO" if reciente < antigua else "MEJORA")  ## Indicamos la dirección de la tendencia observada


# Separamos las cohortes según el periodo en que nacieron.
## P0 corresponde a cohortes nacidas antes de diciembre de 2010.
## P1 corresponde a cohortes nacidas desde diciembre de 2010.

import datetime  ## Importamos datetime para poder comparar las fechas de las cohortes

g0 = ret6.filter(pl.col("cohorte") < datetime.date(2010, 12, 1))  ## Seleccionamos las cohortes nacidas durante P0
g1 = ret6.filter(pl.col("cohorte") >= datetime.date(2010, 12, 1))  ## Seleccionamos las cohortes nacidas durante P1


# Comparamos directamente la retención promedio del sexto mes entre P0 y P1.
## Esta comparación permite evaluar si las cohortes del segundo periodo presentan menor retención que las del primero.

print(f"\nCohortes nacidas en P0 ({g0.height}): {g0['retencion_mes_6'].mean():.2f} %")  ## Mostramos la retención promedio de las cohortes nacidas en P0
print(f"Cohortes nacidas en P1 ({g1.height}) : {g1['retencion_mes_6'].mean():.2f} %")  ## Mostramos la retención promedio de las cohortes nacidas en P1
```

## Ejercicio 4 — Consulta de mercados prioritarios con condiciones múltiples en DuckDB

**Se solicita:** construir una consulta SQL con `SELECT`, `GROUP BY`, `HAVING` y `ORDER BY`; calcular por país los ingresos de cada periodo, la variación absoluta y la cantidad de clientes distintos; conservar solo los mercados con variación negativa, ingresos de P0 sobre un umbral declarado y más de una cantidad mínima de clientes; ordenar por variación ascendente.

```python
# ===== EJERCICIO 4 =====
UMBRAL_INGRESO  = 20_000
MINIMO_CLIENTES = 5

prioritarios = duckdb.sql(f"""
    SELECT  pais,
            ROUND(SUM(CASE WHEN periodo = 'P0' THEN ingreso ELSE 0 END), 2) AS p0,
            ROUND(SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE 0 END), 2) AS p1,
            ROUND(SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE -ingreso END), 2) AS variacion,
            ROUND(100 * SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE -ingreso END)
                      / SUM(CASE WHEN periodo = 'P0' THEN ingreso ELSE 0 END), 2) AS variacion_pct,
            COUNT(DISTINCT cliente_id) AS clientes
    FROM    comparables
    GROUP BY pais
    HAVING  SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE -ingreso END) < 0
      AND   SUM(CASE WHEN periodo = 'P0' THEN ingreso ELSE 0 END) > {UMBRAL_INGRESO}
      AND   COUNT(DISTINCT cliente_id) > {MINIMO_CLIENTES}
    ORDER BY variacion ASC
""").pl()

print(f"Criterios declarados:")
print(f"  1. variacion < 0")
print(f"  2. ingresos P0 > £ {UMBRAL_INGRESO:,}")
print(f"  3. clientes identificados > {MINIMO_CLIENTES}\n")
print(prioritarios)
print(f"\nMercados que cumplen los tres criterios: {prioritarios.height}")
if prioritarios.height:
    top = prioritarios.row(0, named=True)
    print(f"Encabeza el ordenamiento: {top['pais']} "
          f"(£ {top['variacion']:,.2f}, {top['variacion_pct']:.2f} %, {top['clientes']} clientes)")

descartados = duckdb.sql(f"""
    SELECT  pais,
            ROUND(SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE -ingreso END), 2) AS variacion,
            COUNT(DISTINCT cliente_id) AS clientes
    FROM    comparables GROUP BY pais
    HAVING  SUM(CASE WHEN periodo = 'P1' THEN ingreso ELSE -ingreso END) < 0
      AND   SUM(CASE WHEN periodo = 'P0' THEN ingreso ELSE 0 END) > {UMBRAL_INGRESO}
      AND   COUNT(DISTINCT cliente_id) <= {MINIMO_CLIENTES}
    ORDER BY variacion ASC
""").pl()
print("\nMercados grandes descartados por concentracion de clientes:")
print(descartados)
```

### **Mercados prioritarios según condiciones múltiples**

La consulta seleccionó **3 mercados prioritarios** que cumplen simultáneamente los tres criterios: presentar una variación negativa, superar **£20,000 de ingresos en P0** y tener **más de 5 clientes identificados**. Los mercados son **Dinamarca, Suecia y Channel Islands**.

**Dinamarca** encabeza el ordenamiento con una caída de **−£31,150.91 (−63.30%)** y **11 clientes**. Le sigue **Suecia**, con **−£12,899.48 (−26.06%)** y **19 clientes**, y finalmente **Channel Islands**, con **−£4,233.65 (−17.62%)** y **13 clientes**.

Un resultado importante es que **EIRE queda descartado**, a pesar de tener la mayor caída absoluta de todos los mercados con **−£88,853.33**, porque solo tiene **3 clientes identificados** y la consulta exige más de 5. Esto demuestra que el criterio de clientes cambia la priorización: EIRE aparece como un problema de concentración de cuentas, mientras que Dinamarca, Suecia y Channel Islands cumplen las condiciones para ser considerados problemas de mercado.

En conclusión, bajo estos criterios, **Dinamarca es el mercado prioritario**, porque combina una caída significativa en libras y porcentaje con una cantidad de clientes suficiente para justificar un análisis de mercado.

1. **La naturaleza del problema difiere.** EIRE pierde el 25 % de su facturación repartida entre **3 clientes identificados**: basta con que una cuenta mayorista haya reducido pedidos para explicar los £ 88 853. Dinamarca pierde el 63 % repartido entre **11 clientes**: que once cuentas se contraigan a la vez en dos tercios no se explica por una decisión individual, sino por algo que cambió en la plaza.
2. **La probabilidad de recuperación difiere.** Un cliente que reduce pedidos pero sigue comprando —EIRE mantiene 272 facturas frente a 299— **sigue siendo cliente**, y recuperar volumen con quien ya compra es la operación comercial más barata que existe. Un mercado que perdió dos tercios de su facturación puede haber cambiado de proveedor, y recuperarlo exige una oferta nueva, no una llamada.
3. **El riesgo de no actuar difiere.** En EIRE el riesgo está acotado y es visible: si la cuenta se pierde del todo, se pierden los £ 263 710 restantes, pero se sabrá de inmediato. En Dinamarca el riesgo es de contagio: si el motivo del retroceso es un competidor nórdico, lo mismo puede estar empezando en Suecia —que ya cae 26.06 %— y en Noruega. **El caso danés es un caso de aprendizaje sobre una región; el irlandés es un caso de gestión sobre una cuenta.**
4. **El costo de la acción difiere.** Reactivar tres cuentas irlandesas es una tarea de días para el equipo existente. Diagnosticar y recuperar un mercado nórdico es un proyecto de meses con presupuesto propio. Ante recursos limitados, la secuencia racional es **hacer primero lo barato y luego lo caro**, pero asignar el recurso escaso —el analista, el presupuesto de investigación de mercado— al caso danés.

### 3. La limitación del análisis: qué NO demuestra este trabajo

**Este análisis reconcilia exactamente y, aun así, no demuestra ninguna causa.** La reconciliación prueba que la aritmética es correcta y completa, no que los factores identificados hayan producido el resultado. Las ocho limitaciones que deben constar por escrito en el informe son:

1. **No demuestra causalidad.** El puente es una identidad contable. Que el volumen explique £ 1.57 millones no significa que «la caída de volumen causó» nada: es el nombre que se le da a una porción de la aritmética.
2. **No resuelve la dirección de la relación con las discontinuaciones.** ¿Se retiraron los productos porque se vendían poco, o se vendieron poco porque se retiraron? El dato es idéntico bajo ambas hipótesis.
3. **No distingue demanda de disponibilidad.** No hay información de stock. La caída de 925 411 unidades puede ser enteramente un problema de abastecimiento, y en ese caso toda recomendación comercial sería la respuesta equivocada al problema correcto.
4. **No mide margen.** Todo el análisis es de ingreso. Un recambio de catálogo hacia productos más caros puede ser rentable o ruinoso según su costo, y el conjunto no contiene costos.
5. **No cubre al 22.77 % de las líneas en el análisis de clientes.** Las cohortes describen a los 5 852 clientes identificados; las conclusiones sobre retención no se extienden automáticamente al total de compradores.
6. **Depende de las reglas de calidad declaradas.** Cuatro reglas descartaron 64 031 filas (6.00 %). Reglas distintas —por ejemplo, restar las cancelaciones en lugar de excluirlas— producirían cifras distintas sin que nadie se haya equivocado.

8. **No incorpora contexto externo.** No hay competencia, macroeconomía, tipo de cambio ni estacionalidad modelada. El periodo 2010-2011 corresponde además a la salida de la crisis financiera y a una recesión en Irlanda, factores que el dato no observa y que ninguna conclusión puede ignorar sin decirlo.

> **Frase de cierre recomendada para la sustentación.** «Este análisis identifica **dónde** está el movimiento y **cuánto** vale cada componente, con reconciliación exacta y trazabilidad verificable de la fuente. **No identifica por qué ocurrió.** Para eso se requiere incorporar stock, costos, calendario de decisiones de catálogo y canal de captación, y en última instancia un diseño experimental. # **Ticket de salida**

### ✅ Ticket 1 — ¿Por qué pueden subir los ingresos mientras caen las unidades?

Porque el ingreso no es una función solo de las unidades. Partiendo de la identidad $R = Q\cdot\bar{p}$, donde $\bar{p}=\sum_i s_i p_i$ es el precio medio ponderado, el ingreso puede subir con $Q$ cayendo siempre que $\bar{p}$ suba lo suficiente para compensar. Y $\bar{p}$ puede subir por dos vías completamente distintas: porque suben los precios individuales $p_i$ (efecto precio) o porque cambia la composición $s_i$ hacia productos más caros (efecto mezcla). **Confundir ambas vías es el error que este laboratorio combate**, porque conducen a decisiones opuestas: una sugiere revisar la política de precios, la otra revisar el surtido.

En el caso concreto, los tres efectos actuaron así sobre el catálogo común:

- **Efecto volumen: −£ 1 568 093.05.** Se vendieron 925 411 unidades menos (−18.04 %). Valorizadas al precio medio de P0 (£ 1.694483), esa pérdida costó £ 1.57 millones. **Empuja el ingreso hacia abajo y es el mayor movimiento adverso del año.**
- **Efecto mezcla: +£ 577 550.82.** La composición se desplazó hacia productos de mayor precio unitario: con la mezcla de P1 y los precios de P0, el precio medio habría sido £ 1.831853 en lugar de £ 1.694483, es decir, **+£ 0.13737 por unidad**. **Empuja hacia arriba.**
- **Efecto precio: −£ 115 918.44.** Los precios efectivos, en promedio, bajaron £ 0.027571 por unidad. **Empuja hacia abajo, y desmiente la hipótesis de que el ticket subió por precios.**

A esos tres se suma el recambio de catálogo: **+£ 2 046 903.92** de referencias nuevas contra **−£ 704 357.16** de referencias retiradas. **El saldo neto de todo es +£ 236 086.10.**

**Respuesta en una frase:** los ingresos subieron **a pesar** de vender menos unidades, porque la empresa vendió una composición de productos más cara y renovó su catálogo con referencias de mayor valor unitario; **no porque subiera los precios, que de hecho bajaron.**

### ✅ Ticket 2 — ¿Qué variable debe mantenerse constante para que la comparación entre dos cohortes sea válida?

**La antigüedad**, es decir, el número de meses transcurridos desde la primera compra ($k$), **no el mes calendario**.

Dos cohortes solo pueden compararse en la misma columna $k$ de la matriz, y únicamente si ambas **efectivamente vivieron** ese mes. Comparar el mes 6 de una cohorte con el mes 2 de otra mezcla dos fenómenos distintos: la maduración natural del cliente y el efecto del calendario. Y comparar contra una celda vacía es peor todavía, porque esa celda significa *aún no observado* y no *abandonaron*: en este laboratorio, 6 de las 25 cohortes no alcanzaron el mes 6 y deben excluirse explícitamente, tal como hace el Ejercicio 3.

Dos exigencias adicionales que un análisis riguroso añade a la principal: **la definición de «activo» debe ser la misma** para ambas cohortes (aquí, al menos una compra en el mes), y **el denominador debe ser siempre el tamaño inicial de la cohorte**, nunca el del mes anterior, porque en ese caso las tasas dejarían de ser comparables entre sí.

### ✅ Ticket 3 — Driver dominante, cifra que lo sustenta y limitación

**Driver dominante:** la **renovación del catálogo**, con **+£ 2 046 903.92** de las 674 referencias nuevas, es el componente de mayor magnitud absoluta (40.8 % del movimiento bruto). Su contrapeso, y el verdadero problema del año, es el **efecto volumen del catálogo común: −£ 1 568 093.05**.

**Cifra que lo sustenta:** el catálogo común cayó **−£ 1 106 460.66 (−12.73 %)** y el saldo del recambio de referencias fue **+£ 1 342 546.76**. La diferencia entre ambos es exactamente la variación neta observada: **+£ 236 086.10**, con reconciliación verificada a **£ 0.00** de diferencia.

**Limitación:** el análisis **no demuestra causalidad**. Reconcilia, localiza y cuantifica, pero no dispone de stock, costos, motivo de cada discontinuación, canal de captación ni grupo de control. No puede descartarse que la caída de volumen provenga de rotura de stock ni que las discontinuaciones sean consecuencia —y no causa— de la menor demanda. Además, las cohortes describen únicamente a los 5 852 clientes identificados, el 77.23 % de las líneas.

# **Ticket de salida**

### ✅ Ticket 1 — ¿Por qué pueden subir los ingresos mientras caen las unidades?

Los ingresos pueden subir aunque se vendan menos unidades porque también depende del precio medio de los productos vendidos. Este precio puede aumentar por un cambio en los precios o porque se venden más productos de mayor precio. A esto último se le conoce como efecto mezcla.

En nuestro caso, las unidades del catálogo común bajaron **925,411 (−18.04 %)** y el efecto volumen fue de **−£1,568,093.05**. En cambio, el efecto mezcla fue positivo, con **+£577,550.82**, porque la composición de productos vendidos cambió hacia productos de mayor precio. El efecto precio fue negativo, **−£115,918.44**, por lo que los ingresos no aumentaron por una subida general de precios.

Además, los productos nuevos aportaron **+£2,046,903.92**, mientras que los productos discontinuados restaron **−£704,357.16**. Finalmente, los ingresos aumentaron **£236,086.10 (+2.51 %)**.

En resumen, aunque se vendieron menos unidades, el cambio en la mezcla de productos y principalmente las nuevas referencias compensaron la caída del volumen.

### ✅ Ticket 2 — ¿Qué variable debe mantenerse constante para que la comparación entre dos cohortes sea válida?

La variable que debe mantenerse constante es la **antigüedad de la cohorte**, es decir, los meses que han pasado desde la primera compra del cliente.

Por ejemplo, para comparar dos cohortes correctamente debemos comparar su mes 1 con el mes 1 de la otra cohorte, su mes 2 con el mes 2, y así sucesivamente. No sería correcto comparar el mes 6 de una cohorte con el mes 2 de otra porque los clientes tienen diferente tiempo desde su primera compra.

También debemos utilizar la misma definición de cliente activo y el mismo denominador. En nuestro análisis, un cliente es activo si realizó al menos una compra durante el mes y la retención se calcula tomando como referencia el tamaño inicial de la cohorte.

Las celdas vacías tampoco significan que la retención sea 0 %. Significan que la cohorte todavía no ha llegado a esa antigüedad. Por eso, en el análisis del mes 6 se excluyeron las cohortes que todavía no tenían seis meses de observación.

### ✅ Ticket 3 — Driver dominante, cifra que lo sustenta y limitación

En el puente completo, el componente con mayor magnitud fue **Productos nuevos**, con **+£2,046,903.92** de 674 referencias nuevas. Sin embargo, si consideramos solamente los efectos de volumen, mezcla y precio, el principal fue **volumen**, con **−£1,568,093.05**.

El catálogo común disminuyó **−£1,106,460.66 (−12.73 %)**, mientras que el recambio de productos tuvo un saldo positivo de **+£1,342,546.76**. Al compensarse estos movimientos, la variación total terminó siendo de **+£236,086.10** y la reconciliación fue de **£0.00**.

La principal limitación es que este análisis **no demuestra causalidad**. Podemos saber cuánto aportó cada componente a la variación, pero no exactamente por qué ocurrió. Por ejemplo, no tenemos información suficiente sobre stock, costos o motivos de discontinuación para saber si la caída de volumen se debió a menor demanda o a problemas de disponibilidad.

Además, el análisis de cohortes considera a los **5,852 clientes identificados**, por lo que sus resultados de retención no se pueden aplicar automáticamente a todas las líneas del dataset.
