## **Paso 3 — Auditoría inicial y limpieza documentada**


```python
# Prepara las columnas del dataset con los tipos de datos necesarios para el análisis
df = df.with_columns([
    pl.col("InvoiceNo").cast(pl.Utf8),  ## Convierte el número de factura a texto
    pl.col("CustomerID").cast(pl.Utf8),  ## Convierte el identificador del cliente a texto
    pl.col("Quantity").cast(pl.Float64, strict=False),  ## Convierte la cantidad a número decimal
    pl.col("UnitPrice").cast(pl.Float64, strict=False),  ## Convierte el precio unitario a número decimal
    pl.col("InvoiceDate").cast(pl.Utf8)  ## Convierte inicialmente la fecha a texto
      .str.to_datetime(format=FORMATO_FECHA, strict=False)  ## Convierte el texto al formato fecha-hora definido
      .alias("InvoiceDate"),  ## Mantiene el nombre original de la columna
])

# Crea variables necesarias para identificar cancelaciones y calcular el valor de cada línea
df = df.with_columns([
    pl.col("InvoiceNo").str.to_uppercase()  ## Convierte las facturas a mayúsculas
      .str.starts_with("C")  ## Comprueba si la factura comienza con "C"
      .alias("es_cancelacion"),  ## Guarda el resultado como indicador de cancelación

    (pl.col("Quantity") * pl.col("UnitPrice")).alias("importe_linea"),  ## Calcula el importe de cada línea
])

# Verifica que las fechas se hayan convertido correctamente
nulas = df["InvoiceDate"].null_count()  ## Cuenta las fechas que quedaron como valores nulos

print("Fechas que no pudieron parsearse:", nulas)  ## Muestra el resultado del control

# Detiene el proceso si alguna fecha no pudo interpretarse correctamente
if nulas:
    raise ValueError("El formato declarado en FORMATO_FECHA no corresponde a la fuente.")

# Define los valores esperados para comprobar que la fuente original no haya cambiado
FILAS_ESPERADAS = 541_909  ## Cantidad de filas esperada
FACTURAS_ESPERADAS = 25_900  ## Cantidad de facturas distintas esperada

# Verifica que el dataset tenga la cantidad de filas esperada
if df.height != FILAS_ESPERADAS:  ## Compara las filas actuales con las filas esperadas
    raise ValueError(  ## Detiene el código si las cantidades no coinciden
        f"La fuente cambio: se esperaban {FILAS_ESPERADAS:,} filas y llegaron {df.height:,}. "  ## Muestra los valores esperados y encontrados
        "Avise al docente antes de continuar."  ## Indica qué hacer si la fuente cambió
    )

# Verifica que el dataset tenga la cantidad de facturas esperada
if df["InvoiceNo"].n_unique() != FACTURAS_ESPERADAS:  ## Cuenta las facturas diferentes y las compara con las esperadas
    raise ValueError(  ## Detiene el código si las cantidades no coinciden
        f"La fuente cambio: se esperaban {FACTURAS_ESPERADAS:,} facturas distintas "  ## Muestra la cantidad de facturas esperada
        f"y llegaron {df['InvoiceNo'].n_unique():,}."  ## Muestra la cantidad de facturas encontrada
    )

# Confirma que los controles de integridad fueron correctos
print(
    "Integridad verificada:",  ## Indica que las comprobaciones fueron exitosas
    f"{df.height:,}",  ## Muestra la cantidad de filas encontradas
    "filas y",
    f"{df['InvoiceNo'].n_unique():,}",  ## Muestra la cantidad de facturas diferentes
    "facturas, como se esperaba."  ## Confirma que los valores coinciden con la fuente esperada
)

# Construye una tabla resumen para controlar la calidad de los datos
control = pl.DataFrame({
    "indicador": [
        "filas",  ## Cantidad total de registros
        "facturas",  ## Cantidad de facturas distintas
        "clientes",  ## Cantidad de clientes identificados
        "cancelaciones",  ## Cantidad de registros identificados como cancelaciones
        "cantidades_no_positivas",  ## Registros con cantidades menores o iguales a cero
        "precios_no_positivos"  ## Registros con precios menores o iguales a cero
    ],

    "valor": [
        df.height,  ## Obtiene el número total de filas
        df["InvoiceNo"].n_unique(),  ## Cuenta las facturas diferentes
        df["CustomerID"].drop_nulls().n_unique(),  ## Cuenta clientes excluyendo valores nulos
        int(df["es_cancelacion"].sum()),  ## Suma las cancelaciones identificadas
        int((df["Quantity"] <= 0).sum()),  ## Cuenta cantidades no positivas
        int((df["UnitPrice"] <= 0).sum())  ## Cuenta precios no positivos
    ],
})

# Muestra el resumen de los principales controles de calidad
display(control)

# Obtiene y muestra la primera y última fecha registrada en el dataset
print(
    "Rango:",
    df["InvoiceDate"].min(),
    "->",
    df["InvoiceDate"].max()
)
```

## Paso 4 — Definición del periodo comparable

La fuente empieza el 1/12/2010 y termina el 9/12/2011. Para no comparar meses incompletos con meses completos, el laboratorio informa sobre **enero–noviembre de 2011**.

**Cuidado con el sesgo de ventana.** La tabla de facturas se construye sobre *todo* el historial disponible, y el recorte a la ventana se aplica **después** de determinar la recurrencia. Si se recorta antes, un cliente que compró en diciembre de 2010 y volvió en enero de 2011 aparece como comprador nuevo, y la North Star crece de forma artificial en los primeros meses: en esta base, enero pasa de 570 a 246 compras recurrentes, un 57 % menos, solo por haber filtrado en el orden equivocado.


```python
# Crea una columna con el año y mes de cada compra
compras = compras.with_columns(  ## Agrega una nueva columna al DataFrame
    pl.col("InvoiceDate").dt.strftime("%Y-%m").alias("mes")  ## Extrae año-mes de la fecha y lo guarda como "mes"
)

# Define el inicio y fin de la ventana temporal comparable
INICIO, FIN = pl.datetime(2011, 1, 1), pl.datetime(2011, 12, 1)  ## Analiza el periodo de enero a noviembre de 2011

# Agrupa las líneas de productos para obtener una fila por factura
facturas = (
    compras.group_by(["InvoiceNo", "CustomerID", "mes"])  ## Agrupa por factura, cliente y mes
    .agg([
        pl.col("InvoiceDate").min().alias("fecha_factura"),  ## Obtiene la primera fecha registrada de cada factura
        pl.col("importe_linea").sum().alias("importe_factura"),  ## Suma el importe de todos los productos de la factura
        pl.col("Quantity").sum().alias("unidades"),  ## Calcula el total de unidades compradas
        pl.col("StockCode").count().alias("lineas")  ## Cuenta cuántas líneas de productos tiene la factura
    ])
)

# Selecciona las facturas que pertenecen a la ventana comparable
en_ventana = facturas.filter(
    (pl.col("fecha_factura") >= INICIO) &  ## Incluye facturas desde el 1 de enero de 2011
    (pl.col("fecha_factura") < FIN)  ## Excluye el 1 de diciembre y posteriores
)

# Muestra la cantidad de facturas de todo el historial
print("Facturas en el historial completo :", facturas["InvoiceNo"].n_unique())  ## Cuenta las facturas diferentes del historial

# Muestra la cantidad de facturas dentro de la ventana comparable
print("Facturas en la ventana comparable :", en_ventana["InvoiceNo"].n_unique())  ## Cuenta las facturas del periodo seleccionado

# Muestra la cantidad de clientes presentes en la ventana
print("Clientes en la ventana            :", en_ventana["CustomerID"].n_unique())  ## Cuenta los clientes diferentes del periodo

# Calcula los ingresos totales dentro de la ventana
print("Ingresos en la ventana (GBP)      :", round(en_ventana["importe_factura"].sum(), 2))  ## Suma los importes y redondea a 2 decimales

# Muestra las primeras filas para comprobar la información agrupada
display(facturas.head())  ## Permite revisar el resultado de la agrupación
```

### **Facturación y clientes en la ventana comparable**

La agrupación de las transacciones permitió obtener **18 532 facturas en todo el historial**. De ellas, **16 354 facturas** corresponden a la ventana comparable de enero a noviembre de 2011.

Durante este periodo se identificaron **4 173 clientes** y se generaron **GBP 7 820 501,22 en ingresos**. Además, cada factura quedó resumida con su importe total, unidades y cantidad de líneas de productos, facilitando el cálculo de las métricas posteriores.

## Paso 5 — Recurrencia y North Star

**Regla operativa del laboratorio:** un cliente se considera recurrente desde su segunda factura válida, contada sobre **todo el historial disponible**. Su primera compra no se reclasifica retrospectivamente.

Enero de 2011 conserva un sesgo residual, porque solo dispone de un mes previo de historial. Por eso se marca como **mes de calentamiento** y se excluye de las comparaciones de variación mensual.


```python
# Ordena las facturas por cliente, fecha y número de factura
facturas = facturas.sort(["CustomerID", "fecha_factura", "InvoiceNo"])  ## Permite establecer el orden cronológico de las compras de cada cliente

# Numera las compras de cada cliente comenzando desde 1
facturas = facturas.with_columns(
    (pl.int_range(pl.len()).over("CustomerID") + 1).alias("n_compra_cliente")  ## Asigna un número consecutivo a cada compra del cliente
)

# Identifica como recurrente toda compra realizada a partir de la segunda compra
facturas = facturas.with_columns(
    (pl.col("n_compra_cliente") >= 2).alias("es_recurrente")  ## La primera compra no es recurrente; desde la segunda sí
)

# Recorta el análisis a la ventana comparable después de determinar la recurrencia
facturas = facturas.filter(
    (pl.col("fecha_factura") >= INICIO) &  ## Incluye las facturas desde enero de 2011
    (pl.col("fecha_factura") < FIN)  ## Excluye las facturas desde diciembre de 2011
)

# Define enero como mes de calentamiento porque no tiene un mes previo comparable
MES_CALENTAMIENTO = "2011-01"  ## Se utilizará para evitar comparaciones incompletas en enero

# Calcula las métricas generales de cada mes
mensual = (
    facturas.group_by("mes")  ## Agrupa las facturas por mes
    .agg([
        pl.col("InvoiceNo").n_unique().alias("facturas_validas"),  ## Cuenta las facturas válidas diferentes
        pl.col("CustomerID").n_unique().alias("clientes_activos"),  ## Cuenta los clientes activos diferentes
        pl.col("importe_factura").sum().alias("ingresos")  ## Suma los ingresos de cada mes
    ])
)

# Calcula las métricas correspondientes únicamente a las compras recurrentes
rec = (
    facturas.filter("es_recurrente")  ## Conserva solo las compras realizadas por clientes recurrentes
    .group_by("mes")  ## Agrupa las compras recurrentes por mes
    .agg([
        pl.col("InvoiceNo").n_unique().alias("compras_recurrentes"),  ## Cuenta las facturas recurrentes
        pl.col("CustomerID").n_unique().alias("clientes_recurrentes"),  ## Cuenta los clientes recurrentes
        pl.col("importe_factura").sum().alias("ingresos_recurrentes")  ## Suma los ingresos generados por recurrentes
    ])
)

# Une las métricas generales y recurrentes en una sola tabla
kpi = mensual.join(rec, on="mes", how="left").fill_null(0).sort("mes")  ## Completa con 0 los meses sin compras recurrentes y ordena cronológicamente

# Calcula los KPI relacionados con el comportamiento recurrente
kpi = kpi.with_columns([
    (pl.col("compras_recurrentes") / pl.col("clientes_recurrentes")).alias("frecuencia_recurrente"),  ## Compras recurrentes promedio por cliente recurrente
    (100 * pl.col("ingresos_recurrentes") / pl.col("ingresos")).alias("participacion_ingreso_recurrente_pct"),  ## Porcentaje de ingresos provenientes de clientes recurrentes
])

# Muestra la tabla final de KPI mensuales
display(kpi)  ## Permite revisar las métricas calculadas para cada mes
```

### **Evolución mensual de clientes e ingresos recurrentes**

- **Enero 2011:** Se registraron **987 facturas válidas**, 741 clientes activos y **GBP 569 445,04** en ingresos. Los clientes recurrentes generaron **GBP 296 614,01**, equivalente al **52,09 %** de los ingresos.
- **Febrero 2011:** La participación del ingreso recurrente aumentó a **67,08 %**, con 412 clientes recurrentes.

- **Julio 2011:** Se alcanzó una participación de **88,67 %** de ingresos recurrentes.

## Paso 6 — Árbol de métricas

El árbol es una **hipótesis de gestión**, no una demostración causal.

**Objetivo → North Star → drivers → guardrails**

- Objetivo: incrementar valor recurrente sin deteriorar calidad.
- North Star: compras válidas de clientes recurrentes / mes.
- Driver 1: clientes recurrentes activos.
- Driver 2: frecuencia de compra por recurrente.
- Guardrail 1: tasa de cancelación.
- Guardrail 2: concentración de ingresos en Top 10 clientes.


```python
# Reconstruye la North Star a partir de sus componentes principales
kpi = kpi.with_columns(
    (pl.col("clientes_recurrentes") * pl.col("frecuencia_recurrente")).alias("ns_reconstruida")
    ## Clientes recurrentes × frecuencia recurrente = compras recurrentes esperadas
)

# Calcula la diferencia entre las compras recurrentes originales y las reconstruidas
kpi = kpi.with_columns(
    (pl.col("compras_recurrentes") - pl.col("ns_reconstruida")).alias("error_reconstruccion")
    ## Si el resultado es 0, la relación matemática se cumple correctamente
)

# Muestra las variables necesarias para comprobar la relación
display(
    kpi.select([
        "mes",  ## Periodo analizado
        "compras_recurrentes",  ## Valor observado de la North Star
        "clientes_recurrentes",  ## Cantidad de clientes recurrentes
        "frecuencia_recurrente",  ## Compras promedio por cliente recurrente
        "error_reconstruccion"  ## Diferencia entre el valor observado y el reconstruido
    ])
)
```

### **Consistencia de la descomposición de la North Star**

Los resultados muestran que la relación entre **clientes recurrentes × frecuencia recurrente = compras recurrentes** se cumple correctamente en todos los meses analizados.

En **10 de los 11 meses**, el `error_reconstruccion` es exactamente **0.0**. En abril aparece un valor de **1,1369 × 10⁻¹³**, que es prácticamente cero y corresponde a una diferencia mínima causada por la precisión de los cálculos decimales.

Esto confirma que las **compras recurrentes**, utilizadas como North Star, pueden descomponerse matemáticamente en dos componentes: **cantidad de clientes recurrentes** y **frecuencia de compra**.

**Conclusión:** la reconstrucción es consistente y permite utilizar **clientes recurrentes** y **frecuencia recurrente** como drivers de la North Star.

## Paso 7 — Guardrail 1: tasa de cancelación


```python
# Selecciona las transacciones del periodo comparable con información necesaria
base_total = (
    df.filter(
        pl.col("InvoiceDate").is_not_null()  ## Conserva registros con fecha válida
        & pl.col("CustomerID").is_not_null()  ## Conserva registros con cliente identificado
        & (pl.col("InvoiceDate") >= INICIO)  ## Incluye registros desde enero de 2011
        & (pl.col("InvoiceDate") < FIN)  ## Excluye registros desde diciembre de 2011
    )
    .with_columns(
        pl.col("InvoiceDate").dt.strftime("%Y-%m").alias("mes")  ## Crea el periodo año-mes
    )
)

# Obtiene una fila única por factura, cliente y mes
inv_total = base_total.select(
    ["mes", "InvoiceNo", "CustomerID", "es_cancelacion"]
).unique()  ## Evita contar varias veces una misma factura

# Agrupa las facturas por mes para calcular las cancelaciones
cancel = (
    inv_total.group_by("mes")  ## Agrupa los registros según el mes
    .agg([
        pl.col("InvoiceNo").n_unique().alias("facturas_total"),  ## Cuenta todas las facturas diferentes
        pl.col("es_cancelacion").sum().alias("facturas_canceladas")  ## Cuenta las facturas identificadas como canceladas
    ])

    # Calcula el porcentaje de facturas canceladas sobre el total
    .with_columns(
        (
            100 * pl.col("facturas_canceladas") / pl.col("facturas_total")
        ).alias("tasa_cancelacion_pct")  ## Tasa de cancelación expresada en porcentaje
    )

    .sort("mes")  ## Ordena los resultados cronológicamente
)

# Muestra la tasa de cancelación mensual
display(cancel)  ## Presenta el resumen para su análisis
```

### **Comportamiento de la tasa de cancelación**

La tasa de cancelación presenta una **tendencia general a la baja** durante el periodo analizado. En enero de 2011 fue de **20,15 %**, siendo el valor más alto, mientras que en noviembre disminuyó hasta **13,87 %**, el valor más bajo del periodo.

Aunque existen algunas variaciones entre meses, la reducción progresiva indica que las cancelaciones representan una proporción cada vez menor del total de facturas. Por ejemplo, en mayo fue de **15,90 %**, en septiembre de **15,50 %** y en octubre de **14,76 %**.

Al mismo tiempo, el número total de facturas aumentó de **1 236 en enero a 3 086 en noviembre**, mientras que las facturas canceladas pasaron de **249 a 428**. Esto significa que las cancelaciones aumentaron en cantidad absoluta, pero **crecieron a un ritmo menor que el total de facturas**.

**Conclusión:** la tasa de cancelación muestra una mejora durante 2011, lo que constituye un indicador útil para vigilar la calidad del crecimiento y evitar que el aumento de ventas esté acompañado por un nivel excesivo de cancelaciones.

## Paso 8 — Guardrail 2: concentración Top 10 clientes


```python
# Calcula el ingreso total generado por cada cliente en cada mes
cliente_mes = (
    facturas.group_by(["mes", "CustomerID"])  ## Agrupa por mes y cliente
    .agg(
        pl.col("importe_factura").sum().alias("ingreso_cliente")  ## Suma el ingreso generado por cada cliente
    )
)

# Ordena los clientes de mayor a menor ingreso dentro de cada mes
ordenado = cliente_mes.sort(
    ["mes", "ingreso_cliente"],  ## Ordena primero por mes y luego por ingreso
    descending=[False, True]  ## Mes ascendente e ingresos descendentes
)

# Asigna una posición a cada cliente según su ingreso mensual
ordenado = ordenado.with_columns(
    (pl.int_range(pl.len()).over("mes") + 1).alias("ranking")  ## El cliente con mayor ingreso recibe el ranking 1
)

# Calcula cuánto representan los 10 clientes con mayores ingresos de cada mes
conc = (
    ordenado.group_by("mes")  ## Agrupa nuevamente los resultados por mes
    .agg([
        pl.col("ingreso_cliente").sum().alias("ingreso_mes"),  ## Calcula el ingreso total del mes

        pl.col("ingreso_cliente")
          .filter(pl.col("ranking") <= 10)  ## Selecciona únicamente los 10 clientes con mayor ingreso
          .sum()
          .alias("ingreso_top10")  ## Suma los ingresos generados por esos 10 clientes
    ])

    # Calcula el porcentaje del ingreso mensual concentrado en los 10 principales clientes
    .with_columns(
        (
            100 * pl.col("ingreso_top10") / pl.col("ingreso_mes")
        ).alias("concentracion_top10_pct")  ## Expresa la concentración como porcentaje
    )

    .sort("mes")  ## Ordena los resultados cronológicamente
)

# Muestra la concentración de ingresos de los principales clientes
display(conc)  ## Permite analizar la dependencia de los 10 clientes principales
```

### **Concentración de ingresos en los 10 principales clientes**

Los resultados muestran que la concentración de ingresos en los **10 clientes principales varía considerablemente** durante el periodo. En enero alcanza su nivel más alto, con **33,94 %**, lo que significa que aproximadamente un tercio de los ingresos del mes provino de solo 10 clientes.

A partir de febrero, la concentración disminuye hasta **20,05 %** y continúa variando durante el año. En junio vuelve a aumentar hasta **29,22 %**, mientras que en noviembre alcanza uno de sus niveles más bajos, con **15,66 %**, a pesar de que los ingresos totales fueron los más altos del periodo (**GBP 1 161 800 aproximadamente**).

Esto indica que el crecimiento de los ingresos hacia noviembre **no depende proporcionalmente de los 10 clientes principales**, sino que los ingresos están más distribuidos entre otros clientes.

**Conclusión:** la concentración Top 10 funciona como un **guardrail**, porque permite vigilar el riesgo de depender demasiado de pocos clientes. En este caso, la reducción de **33,94 % en enero a 15,66 % en noviembre** sugiere una menor dependencia relativa de los principales clientes.

```python
# Integra los guardrails al tablero principal de KPI
kpi = (
    kpi
    .join(
        cancel.select(["mes", "tasa_cancelacion_pct"]),  ## Agrega la tasa de cancelación por mes
        on="mes",  ## Utiliza el mes como clave para unir las tablas
        how="left"  ## Conserva todos los meses del KPI principal
    )
    .join(
        conc.select(["mes", "concentracion_top10_pct"]),  ## Agrega la concentración de ingresos del Top 10
        on="mes",  ## Une nuevamente utilizando el mes
        how="left"  ## Mantiene todos los registros de la tabla principal
    )
    .sort("mes")  ## Ordena los resultados cronológicamente
)

# Calcula la variación mensual de la North Star y de sus principales drivers
kpi = kpi.with_columns([
    (
        100 * (
            pl.col("compras_recurrentes") /
            pl.col("compras_recurrentes").shift(1) - 1
        )
    ).alias("var_ns_pct"),  ## Mide cuánto aumentaron o disminuyeron las compras recurrentes frente al mes anterior

    (
        100 * (
            pl.col("clientes_recurrentes") /
            pl.col("clientes_recurrentes").shift(1) - 1
        )
    ).alias("var_clientes_rec_pct"),  ## Mide el cambio porcentual mensual de clientes recurrentes

    (
        100 * (
            pl.col("frecuencia_recurrente") /
            pl.col("frecuencia_recurrente").shift(1) - 1
        )
    ).alias("var_frecuencia_pct"),  ## Mide el cambio porcentual mensual en la frecuencia de compra
])

# Muestra el tablero completo con KPI, drivers y guardrails
display(kpi)  ## Permite analizar la evolución mensual y detectar cambios relevantes
```

### **Evolución de la North Star, drivers y guardrails**

Los resultados muestran que las **compras recurrentes**, nuestra North Star, presentan un crecimiento importante durante el periodo. Pasan de **570 en enero a 2 334 en noviembre**, aunque existen algunas disminuciones puntuales, como en abril (-2,30 %), junio (-9,44 %) y agosto (-2,80 %).

Este comportamiento está explicado principalmente por el crecimiento de los **clientes recurrentes**. Por ejemplo, en noviembre aumentaron **32,54 %** respecto a octubre. Además, la **frecuencia recurrente** aumentó **12,09 %**, alcanzando **1,67 compras por cliente**, su valor más alto del periodo.

Por otro lado, los guardrails muestran una evolución favorable. La **tasa de cancelación** disminuyó de **20,15 % en enero a 13,87 % en noviembre**. La **concentración del Top 10** también terminó en un nivel menor, pasando de **33,94 % a 15,66 %**, lo que indica una menor dependencia relativa de los principales clientes.

La `ns_reconstruida` coincide con las `compras_recurrentes` en todos los meses y el `error_reconstruccion` es **0**, salvo una diferencia prácticamente insignificante en abril debido a la precisión decimal. Esto confirma que la North Star está correctamente descompuesta mediante sus drivers.

**Conclusión:** el crecimiento de la North Star hacia noviembre está asociado principalmente con un mayor número de **clientes recurrentes** y, en ese último mes, también con una mayor **frecuencia de compra**, mientras que los guardrails no muestran un deterioro proporcional. Por ello, el crecimiento observado presenta señales de ser **crecimiento útil y no solamente aparente**.

La tasa de recurrencia corrige esa limitación: expresa qué proporción de las facturas del mes corresponde a clientes que ya habían comprado con anterioridad. Se trata de una magnitud relativa y, por tanto, comparable entre meses de distinto volumen.

### **Cálculo de la tasa de recurrencia mensual**

$$
\text{Tasa de recurrencia} = \frac{\text{Facturas recurrentes}}{\text{Facturas totales}} \times 100
$$


La tasa indica qué **porcentaje de las facturas del mes corresponde a clientes que ya habían comprado anteriormente**.

```python
# ---------------------------------------------------------------------------
# CELDA DE TRABAJO — Ejercicio 1
# Desarrolle aquí su código.
# ---------------------------------------------------------------------------

# Calcula las facturas totales y recurrentes por cada mes
recurrencia = (
    facturas.group_by("mes")  ## Agrupa las facturas según el mes
    .agg([
        pl.col("InvoiceNo").n_unique().alias("facturas_totales"),  ## Cuenta las facturas válidas del mes
        pl.col("InvoiceNo").filter(pl.col("es_recurrente")).n_unique().alias("facturas_recurrentes")  ## Cuenta las facturas de clientes recurrentes
    ])
    .with_columns(
        (100 * pl.col("facturas_recurrentes") / pl.col("facturas_totales"))
        .alias("tasa_recurrencia_pct")  ## Calcula qué porcentaje de facturas corresponde a clientes recurrentes
    )
    .sort("mes")  ## Ordena los resultados cronológicamente
)

# Incorpora la tasa de recurrencia a la tabla de KPI del reto
kpi_reto = kpi_reto.join(
    recurrencia.select(["mes", "tasa_recurrencia_pct"]),  ## Selecciona únicamente las columnas necesarias
    on="mes",  ## Utiliza el mes como clave de unión
    how="left"  ## Conserva todos los meses de kpi_reto
)

# Muestra la tasa de recurrencia mensual
display(
    recurrencia  ## Presenta las facturas totales, recurrentes y la tasa calculada
)

# Identifica el mes con la tasa de recurrencia más alta
mayor_recurrencia = recurrencia.sort(
    "tasa_recurrencia_pct", descending=True  ## Ordena de mayor a menor tasa
).head(1)

# Identifica el mes con la tasa de recurrencia más baja
menor_recurrencia = recurrencia.sort(
    "tasa_recurrencia_pct"  ## Ordena de menor a mayor tasa
).head(1)

print("Mes con mayor tasa de recurrencia:")
display(mayor_recurrencia)

print("Mes con menor tasa de recurrencia:")
display(menor_recurrencia)

# Compara el ranking por tasa de recurrencia con el ranking por compras recurrentes
comparacion = recurrencia.select([
    "mes",
    "tasa_recurrencia_pct",
    "facturas_recurrentes"
]).sort("tasa_recurrencia_pct", descending=True)  ## Ordena los meses según la tasa de recurrencia

display(comparacion)
```

El árbol de métricas descompuso la North Star en clientes recurrentes y frecuencia de compra. Ninguno de esos dos drivers recoge el valor económico de cada transacción: dos meses con idéntica cantidad de compras recurrentes pueden presentar ingresos muy distintos si el importe promedio de la factura cambia.

El ticket promedio de la factura recurrente completa esa descripción y permite establecer si el crecimiento observado proviene de mayor actividad o de mayor valor por transacción.

**Se solicita:**

4. Contrastar los meses de mayor caída de la North Star con el comportamiento del ticket promedio en esos mismos meses.

**Tiempo estimado:** 5 minutos.


### **Cálculo del ticket promedio recurrente**

$$
\text{Ticket promedio recurrente} =
\frac{\text{Ingresos recurrentes}}
{\text{Facturas recurrentes}}
$$

```python
# ---------------------------------------------------------------------------
# CELDA DE TRABAJO — Ejercicio 2
# Desarrolle aquí su código.
# ---------------------------------------------------------------------------
# Calcula los ingresos y las facturas recurrentes de cada mes
ticket_recurrente = (
    facturas.filter(
        pl.col("es_recurrente")  ## Considera únicamente las facturas de clientes recurrentes
    )
    .group_by("mes")  ## Agrupa las facturas recurrentes por mes
    .agg([
        pl.col("importe_factura").sum().alias("ingresos_recurrentes"),  ## Suma los ingresos generados por facturas recurrentes
        pl.col("InvoiceNo").n_unique().alias("facturas_recurrentes")  ## Cuenta las facturas recurrentes del mes
    ])
    .with_columns(
        (
            pl.col("ingresos_recurrentes") / pl.col("facturas_recurrentes")
        ).alias("ticket_promedio_recurrente")  ## Calcula el ingreso promedio generado por cada factura recurrente
    )
    .sort("mes")  ## Ordena los meses cronológicamente
)

# Incorpora el ticket promedio recurrente a la tabla de KPI del reto
kpi_reto = kpi_reto.drop("ticket_promedio_recurrente") if "ticket_promedio_recurrente" in kpi_reto.columns else kpi_reto  ## Evita duplicar la columna si se ejecuta nuevamente

kpi_reto = kpi_reto.join(
    ticket_recurrente.select(["mes", "ticket_promedio_recurrente"]),  ## Selecciona el mes y el ticket calculado
    on="mes",  ## Utiliza el mes como clave de unión
    how="left"  ## Conserva todos los meses de kpi_reto
)

# Calcula la variación mensual porcentual del ticket promedio recurrente
kpi_reto = kpi_reto.with_columns(
    (
        100 * (
            pl.col("ticket_promedio_recurrente") /
            pl.col("ticket_promedio_recurrente").shift(1) - 1
        )
    ).alias("var_ticket_promedio_pct")  ## Mide cuánto aumenta o disminuye el ticket respecto al mes anterior
)

# Muestra el ticket promedio y su variación mensual
display(
    kpi_reto.select([
        "mes",  ## Mes analizado
        "compras_recurrentes",  ## North Star del laboratorio
        "ticket_promedio_recurrente",  ## Valor promedio de cada factura recurrente
        "var_ticket_promedio_pct"  ## Variación porcentual respecto al mes anterior
    ])
)
```

### Ejercicio 4 — Consulta de meses saludables con condiciones múltiples en DuckDB

En la Actividad 2 se empleó DuckDB para ordenar los meses según la variación de la North Star. Una consulta orientada a la decisión, sin embargo, no se limita a ordenar: delimita el subconjunto de periodos que satisfacen simultáneamente el objetivo de crecimiento y las restricciones fijadas por los guardrails.

Un mes de crecimiento acompañado de un deterioro en la tasa de cancelación no constituye un mes saludable. Esta consulta materializa esa distinción en una regla reproducible.

3. Ordenar el resultado por variación de la North Star de mayor a menor.
4. Determinar cuántos meses satisfacen ambos criterios.

**Tiempo estimado:** 5 minutos.


### **Criterios para identificar un mes saludable**

$$
\text{Mes saludable} =
(\text{Variación North Star} > 0)
\land
(\text{Tasa de cancelación} < \text{Promedio de cancelación})
$$

```python
# Calcula el promedio de la tasa de cancelación del periodo
promedio_cancelacion = duckdb.sql("""
SELECT AVG(tasa_cancelacion_pct) AS promedio_cancelacion
FROM kpi_reto
""").fetchone()[0]  ## Obtiene el promedio de cancelación para utilizarlo como referencia

print(f"Promedio de cancelación del periodo: {promedio_cancelacion:.2f}%")  ## Muestra el promedio calculado

# Busca los meses que cumplen simultáneamente las condiciones de crecimiento y calidad
meses_saludables = duckdb.sql("""
SELECT
    mes,
    compras_recurrentes,
    ROUND(var_ns_pct, 1) AS var_ns_pct,
    ROUND(tasa_cancelacion_pct, 1) AS tasa_cancelacion_pct
FROM kpi_reto
WHERE var_ns_pct > 0
  AND tasa_cancelacion_pct < (
      SELECT AVG(tasa_cancelacion_pct)
      FROM kpi_reto
  )
ORDER BY var_ns_pct DESC
""").pl()  ## Convierte el resultado de DuckDB nuevamente a Polars

# Muestra los meses que cumplen ambos criterios
display(meses_saludables)  ## Permite revisar los meses considerados saludables

# Cuenta la cantidad de meses saludables
print("Cantidad de meses saludables:", meses_saludables.height)  ## Muestra cuántos meses cumplen las condiciones
```

- **Mayo** tuvo el mayor crecimiento de la North Star (**49.7 %**) con una cancelación de **15.9 %**, mientras que **noviembre** alcanzó **2 334 compras recurrentes** y una cancelación de **13.9 %**.

El tablero construido en la Actividad 2 presenta la North Star y su driver como series temporales paralelas. Esa disposición permite observar la evolución de ambas magnitudes, pero no muestra con claridad si sus variaciones se corresponden entre sí.

Un gráfico de dispersión que enfrente la variación del driver con la variación de la North Star hace visible esa correspondencia: los puntos alineados sobre una tendencia ascendente indican que el driver acompaña el movimiento del resultado, mientras que los puntos dispersos advierten que otros factores intervienen. Corresponde recordar que la correspondencia observada describe una asociación y no acredita una relación causal.

**Se solicita:**

1. Construir con Plotly un gráfico de dispersión que sitúe la variación porcentual de clientes recurrentes en el eje horizontal y la variación porcentual de la North Star en el eje vertical.
2. Identificar cada punto con el mes correspondiente.
3. Rotular los ejes y titular el gráfico de modo que resulte interpretable sin recurrir al código.
4. Excluir de manera explícita el mes de calentamiento, que carece de variación calculable.

**Tiempo estimado:** 5 minutos.


```python
# ---------------------------------------------------------------------------
# CELDA DE TRABAJO — Ejercicio 5
# Desarrolle aquí su código.
# ---------------------------------------------------------------------------
# Filtra los meses con variaciones calculables
dispersion = kpi_reto.filter(
    (pl.col("mes") > MES_CALENTAMIENTO)  ## Excluye enero, que no tiene variación mensual
    & pl.col("var_clientes_rec_pct").is_not_null()  ## Conserva solo meses con variación de clientes recurrentes
    & pl.col("var_ns_pct").is_not_null()  ## Conserva solo meses con variación de la North Star
)

# Convierte los datos a Pandas para utilizarlos con Plotly
dispersion_pd = dispersion.to_pandas()  ## Prepara la tabla para construir el gráfico

# Construye el gráfico de dispersión
fig = px.scatter(
    dispersion_pd,
    x="var_clientes_rec_pct",  ## Coloca la variación de clientes recurrentes en el eje horizontal
    y="var_ns_pct",  ## Coloca la variación de la North Star en el eje vertical
    text="mes",  ## Identifica cada punto con el mes correspondiente
    title="Relación entre la variación de clientes recurrentes y la North Star"
)

# Personaliza las etiquetas para facilitar la interpretación
fig.update_traces(
    textposition="top center"  ## Coloca el nombre del mes encima de cada punto
)

fig.update_layout(
    xaxis_title="Variación de clientes recurrentes (%)",  ## Etiqueta del eje X
    yaxis_title="Variación de la North Star (%)",  ## Etiqueta del eje Y
    hovermode="closest"  ## Muestra información del punto más cercano al pasar el cursor
)

# Muestra el gráfico
fig.show()
```

- **Junio** presenta el comportamiento más desfavorable, con una disminución de **4.4 % en clientes recurrentes** y de **9.4 % en la North Star**.
- **Conclusión:** existe una **asociación positiva** entre ambas variaciones, lo que respalda a los clientes recurrentes como un driver relevante de la North Star, aunque **no demuestra una relación causal**.
